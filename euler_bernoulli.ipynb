# Euler and Bernoulli's equations

{ref}`Mass conservation<section-mass_con>` applied over a control volume of fluid allowed us to relate flow rates (speed, volume flow rate, mass flow rate) to geometry. But how can we evaluate the pressure field? 

Euler's equation allows us to do this, by applying Newton's second law of motion to a fluid element follwing at streamline. 




:::::{hint} 🔥 Applying Newton's law to a fluid element
:class: concept
:class: dropdown

Consider a small element of fluid moving in the $s$-direction - along a streamline - with velocity $u$ and acceleration $a$. 

```{figure} images/euler_dev1.png
:alt: Fluid element following a streamline
:class: bg-none
:width: 85%
:align: center
```

It is a small element, so it has mass $\delta m = \rho \delta s \delta A$. Newton's second law tells us that the total force acting on this element is its mass multiplied by its acceleration

$
\delta F = \delta m\,a = \rho a \delta s \delta A.
$

However, the net force in the streamwise direction will be the sum of all of the forces acting on the element - shown in red. These are 
- The component of weight ($g \delta m$) acting in the streamwise direction

$
\displaystyle g \delta m \cos\theta = g\rho \delta s \delta A \cos\theta = g\rho \delta s \delta A\frac{\partial z}{\partial s}.
$

- The pressure force $p\delta A$ acting on it's trailing surface in the direction of streamline motion.
- The pressure force $(p+\delta p)\delta A$ acting on it's frontal surface against the direction of streamline motion.

Hence, the net streamwise force

$
\displaystyle \delta F = p\delta A - (p+\delta p)\delta A - g\rho \delta s \delta A\frac{\partial z}{\partial s} = - \delta p\delta A - g\rho \delta s \delta A\frac{\partial z}{\partial s}.
$

From Newton's second law, above, equating $\delta F$

$
\displaystyle \rho a \delta s \delta A = - \delta p\delta A - g\rho \delta s \delta A\frac{\partial z}{\partial s}.
$

Dividing through by $\rho \delta s\delta A$, we find

(eqn-euler1)=
$
\displaystyle \frac{1}{\rho}\frac{\delta p}{\delta s} + g\frac{\partial z}{\partial s} + a = 0.
$

::::{dropdown} What is the acceleration $a$?

In general as a fluid particle moves a small distance $\delta s$, its velocity changes with time and space, i.e. $u = u(s,t)$ with

$
\displaystyle \delta u = \frac{\partial u}{\partial s}\delta s + \frac{\partial u}{\partial t}\delta t,
$

and 

$
\displaystyle  \frac{\delta u}{\delta t} = \frac{\partial u}{\partial s}\frac{\delta s}{\delta t} + \frac{\partial u}{\partial t}.
$

:::{tip}What does $\partial$ mean?
:class: dropdown

The sign $\partial$ indcates `partial derivatives' of velocity with respect to distance and time. Where a property is dependent on more than one variable, the partial derivative considers that change in the property with one variable while holding all others constant. 

:::

In the limit $\delta t \rightarrow 0$, we form the differential equation

$
\displaystyle  \frac{\text{d}u}{\text{d}t} = \frac{\partial u}{\partial s}\frac{\text{d}s}{\text{d}t} + \frac{\partial u}{\partial t}.
$

Noting that the speed $u = \text{d}s/\text{d}t$, the acceleration 

$
\displaystyle  \frac{\text{d}u}{\text{d}t} = u\frac{\partial u}{\partial s} + \frac{\partial u}{\partial t}.
$

For a **steady flow** that does not vary with time, $\partial u/\partial t = 0$ and the acceleration 

(eqn-acc)=
$
\displaystyle a = u\frac{\partial u}{\partial s}
$

::::

We can now combine our {ref}`force balance equation<eqn-euler1>` and this {ref}`acceleration expression<eqn-acc>` and take limits.

:::::

This leads to Euler's equation

$
\displaystyle \frac{1}{\rho}\frac{\text{d}p}{\text{d}s} + g\frac{\text{d}z}{\text{d}s} + u\frac{\text{d}u}{\text{d}s} = 0.
$

Or equivalently, multiplying through by $\text{d}s$


$
\displaystyle \frac{\text{d}p}{\rho} + g\text{d}z + u\text{d}u = 0.
$

:::{hint} Notes on Euler's equation

We have made two key assumptions to derive Euler's equation:

It applies to **steady flow** along a **streamline**.

Further, we have considered there are no external forces applied to the fluid element. This means we have assumed
- **no shaft work**, implying no pumps or turbines
- no shear stresses (i.e. friction) was applied to the fluid element $\implies$ the flow is **inviscid**.

:::

:::{hint} Concept: 💭 Thought experiment - What happens for a static fluid? 
:class: concept

Does Euler's equation predict what we already know for a static fluid? 

If we set the fluid velocity $u=0$ in Euler's equation

$
\displaystyle \frac{\text{d}p}{\rho} + g\text{d}z + u\text{d}u = 0 \quad \implies \quad  \frac{\text{d}p}{\rho} + g\text{d}z = 0
$

and 

$
\text{d}p = -\rho g\text{d}z.
$

If we integrate this differential equation we recover the familiar hydrostatic pressure equation

$
p = p_0-\rho g (z=z_0).
$

::::

As you see, to be able to use the Euler equation more generally, we need to integrate the terms, with

$
\displaystyle \int\,\frac{\text{d}p}{\rho} + \int g\,\text{d}z + \int u\,\text{d}u = \int 0\,\text{d}s.
$

The first term 

$
\int\,\frac{\text{d}p}{\rho}
$

can only be integrated if density is constant i.e. the flow is **incompressible**. In which case, we find the energy form of **Bernoulli's equation**

$
\displaystyle \frac{p}{\rho} + gz + \frac{u^2}{2} = \text{constant}
$

where the constant combines all of the constants of the (indefinite) integrations of each term. 

:::{hint} Can you remember all of the assumptions to get here? 
:class: dropdown

- Steady flow
- Along a streamline
- With no shaft work 
- Inviscid - with no frictional shear stresses
- Incompressible - with constant density

:::

In this form each component has the units energy/unit mass. The equation is a statement of the **law of conservation of energy**.

| $\displaystyle\frac{p}{\rho}$ | + | $\displaystyle gz$ | + | $\displaystyle\frac{u^2}{2}$ | = | constant along a streamline |
|:---:|:---:|:---:|:---:|:---:|:---:|:---|
| Pressure energy || Potential energy || Kinetic energy | = | Total energy |

The equation can be divided through by $g$, the gravitational acceleration, to give each term in units of **elevation**, or **head** i.e. a vertical distance. 

| $\displaystyle\frac{p}{\rho g}$ | + | $\displaystyle z$ | + | $\displaystyle\frac{u^2}{2g}$ | = | $H_T$ |
|:---:|:---:|:---:|:---:|:---:|:---:|:---|
| Static / pressure head || Elevation head || Velocity head | = | Total head |

However, when we are working with pressures, we typically want to use the **pressure form** of Bernoulli's equation. Multiplying through the energy form by $\rho$, 

| $\displaystyle p$ | + | $\displaystyle \rho gz$ | + | $\displaystyle\frac{\rho u^2}{2}$ | = | constant along a streamline |
|:---:|:---:|:---:|:---:|:---:|:---:|:---|
| Static pressure  || Elevation pressure || Dynamic pressure | = | Total pressure |

(def_piezo_press)=
:::{tip} Piezometric pressure
:class: dropdown

The **piezometric pressure**, or head, is the sum of the static and elevation terms. 

In pressure form

$
p + \rho gz.
$

In head form

$
\displaystyle \frac{p}{\rho g} + z = H_{PZ}.
$

This is a often a convenient quantity when working with pressure measurements, simplyfying Bernoulli's statement of energy conservation to

*The sum of the piezometric and dynamic pressures/heads is constant along a streamline.*

:::

# Applications of Bernoulli

### Pipe flow

For steady flow of an ideal fluid along a streamline, with no shaft work, total head is constant. A pipe flow provides a nice example of the terms in Bernoulli's equation. We normally consider the flow relative to the pipe centreline, the central streamline, as is usual for 1-D flow. In general each streamline has its own values of static, dynamic and total heads. 

```{figure} images/pipe_head.png
:alt: Representation of head terms in pipe flow
:class: bg-none
:width: 85%
:align: center
```

::::{prf:example} Steady flow through a constriction
:enumerator: 18

Consider the steady flow of ideal fluid ($\mu = 0$, $\rho =$constant) through a constricted pipe, shown below.

```{figure} images/constriction_b.png
:alt: Steady, ideal flow through a constriction
:class: bg-none
:width: 80%
:align: center
```

Determine whether the pressure at point 2 is higher than, lower than, or the same as the pressure at point 1.

:::{admonition} Solution
:class: dropdown

Conservation of mass (volume for constant density fluid) over the control volume intersecting 1 and 2 gives us

$
\dot V_1 = \dot V_2\quad\implies\quad u_1A_1 = u_2A_2.
$

Since $A_2<A_1$ and 

$
\displaystyle u_1 = \frac{A_2}{A_1}u_2,
$

it follws that $u_2>u_1$.

That is, **the fluid flows faster through the constriction**. 

But we are asked about pressures $\implies$ Bernoulli's equation. Applying Bernoulli along the streamline connecting 1 to 2,

$
\displaystyle p_1 + \rho g h + \frac{\rho u_1^2}{2} = p_2 + \rho g h + \frac{\rho u_2^2}{2}.
$

The streamline is horizontal, so there is no change in elevation $h$, and

$
\displaystyle p_1 - p_2 = \frac{\rho}{2}\left(u_2^2-u_1^2\right).
$

With $u_2>u_1$, $u_2^2-u_1^2>0$ and $p_1-p_2>0$

$
\implies\underline{p_1>p_2.}
$

Inside the constriction the flow speed increases due to mass conservation, and that kinetic energy is extracted from the pressure (no potential energy change). The pressure in the constriction is therefore lower. 

::::

::::{prf:example} Tapering pipe
:enumerator: 19

A pipe carrying water tapers from a cross-section of $0.1\,\text{m}^2$ at A to $0.02\,\text{m}^2$ at B. At A the velocity, assumed uniform, is $2.6\,\text{m}/\text{s}$ and the pressure is $50\,\text{kPa}$ gauge.  

If frictional effects are negligible, determine the pressure at B, which is $3\,\text{m}$ above the level of A.

```{figure} images/constr_soln.png
:alt: Tapering pipe
:class: bg-none
:width: 60%
:align: center
```

:::{admonition} Solution
:class: dropdown

First we apply mass conservation across the control volume intersecting A and B (normal to the flow) and everywhere else with no flow through the boundaries. For constant density, this reduces to continuity

$
u_AA_A=u_BA_B,
$

and 

$
\displaystyle u_b = \frac{u_AA_A}{A_B}.
$

Now we can apply Bernoulli along the streamline connecting A and B (here, in pressure form because we are trying to calculate a pressure)

$
\displaystyle p_A + \rho g z_A + \frac{\rho u_A^2}{2} = p_B + \rho g z_B + \frac{\rho u_B^2}{2}.
$

Rearranging for $p_B$,

$
\displaystyle p_B = p_A + \rho g (z_A-z_B) + \frac{\rho (u_A^2-u_B^2)}{2} 
$

and evaluating from the data provided

$
\displaystyle p_B= 50\times 10^3 + 1000\times9.81(0-3) + \frac{1000 (2.6^2-13^2)}{2}
$

$
\implies\underline{p_B = -60.55\,\text{kPa}~\text{gauge}.}
$

### Venturimeter

This is a well-known flow meter consisting of a convergent-divergent pipe or **venturi**. The fluid is restored almost to its original condition by the divergent section as it slows down smoothly.  

Typically, the geometry will be 
- $30^\circ$ included angle in the convergent part 
- $7^\circ$ included angle in the divergent part
- The diameter ratio $d_{\text{throat}}/d_{\text{pipe}} = 1/2$, where `throat' refers to the narrowest part of the section.

:::{tip} Calculating volume flow rates from a venturimeter

Consider the venturimeter shown below. 

```{figure} images/venturi_slope.png
:alt: Generalised venturi meter configuration
:class: bg-none
:width: 80%
:align: center
```

For constant density fluid (incompressible flow), mass conservation relates the flow speeds ($u$) in the pipe and throat to the (known) geometry of the meter described by the cross-section areas $A$

$
\displaystyle u_1A_1 = u_2A_2 \quad\implies\quad u_1 = u_2\frac{A_2}{A_1}.
$

Bernoulli applied along the streamline connecting pipe (1) to throat (2) is


$
\displaystyle p_1 + \rho g z_1 + \frac{\rho u_1^2}{2} = p_2 + \rho g z_2 + \frac{\rho u_2^2}{2}.
$

Rearranging, we have

$
\displaystyle (p_1 + \rho g z_1) - (p_2 + \rho g z_2) = \frac{\rho(u_2^2-u_1^2)}{2}
$

where the terms on the left hand side are the {ref}`piezometric pressures<def_piezo_press>` at 1 and 2 respectively. Typically, this is rewritten in terms of piezometric head difference, dividing through by $\rho g$

$
\displaystyle H_{PZ1}-H_{PZ2} = \Delta H_{PZ} = \frac{(u_2^2-u_1^2)}{2g}.
$

From continuity we can write this in terms of the throat flow speed only

$
\displaystyle \frac{(u_2^2-u_1^2)}{2g} = u_2^2\left(1-\left(\frac{A_2}{A_1}\right)^2\right) = 2g\Delta H_{PZ}
$

$
\displaystyle \implies u_2 = \sqrt{\frac{2g\Delta H_{PZ}}{\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}
$

or the system volume flow rate $\dot V = u_2A_2$

$
\displaystyle \implies \dot V = A_2\sqrt{\frac{2g\Delta H_{PZ}}{\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}
$

:::

:::{hint} Notes on Venturimeters

- This volume flow rate is referred to as the **ideal flow rate** because it is derived based on the assumptions made in deriving Bernoulli's equation. That is steady, incompressible, inviscid flow along a streamline.

- In practice some energy will be lost due to friction and the velocity will not be completely uniform at inlet and throat.
  
- For a measured value of $\Delta H_{PZ}$, the real volume flow rate will be **lower** than that predicted in the ideal equation.  

- To account for this a discharge coefficient is introduced 

$
\displaystyle c_d = \frac{\dot m_{\text{real}}}{\dot m_{\text{ideal}}}= \frac{\rho\dot V_{\text{real}}}{\rho\dot V_{\text{ideal}}} = \frac{\dot V_{\text{real}}}{\dot V_{\text{ideal}}}
$

- Discharge coefficient $c_d$ is determined experimentally by individual calibration of the venturimeter, typically between $0.95$ and $0.99$.

For a venturimeter of known throat geometry, the **real** volume flow rate measured from the piezometric head difference is

$
\displaystyle \dot V_{\text{real}} = c_d\dot V_{\text{ideal}} = c_dA_2\sqrt{\frac{2g\Delta H_{PZ}}{\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}
$

This equation is in the module formula sheet.

:::

::::{prf:example} Horizontal venturimeter
:enumerator: 20

Water flows along a horizontal venturimeter which has inlet and throat areas of $180\,\text{cm}^2$ and $45\,\text{cm}^2$ respectively.  A mercury-under-water {ref}`manometer<sec_manometers>` connected to the inlet and throat sections shows a difference in mercury levels of $10\,\text{cm}$.  The coefficient of discharge is $c_d=0.96$.  Calculate 

1. the pressure difference between inlet and throat
2. the mass flow rate of the water.   

$\rho_w = 1000\,\text{kg}/\text{m}^3$\
$\rho_m = 13600\,\text{kg}/\text{m}^3$

:::{admonition} Solution
:class: dropdown

1. For a {ref}`differential manometer<exp_diff_man>` with reading $\Delta z_m$

$
\Delta p = \Delta z_m (\rho_m-\rho_w)g = 0.1(13600 - 1000)9.81
$

$
\implies\underline{\Delta p = 12.36\,\text{kPa}.}
$

2. The mass flow rate through the meter can be found using the venturimeter discharge equation

$
\displaystyle \dot V_{\text{real}} = c_dA_2\sqrt{\frac{2g\Delta H_{PZ}}{\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}.
$

The difference in piezometric head

$
\displaystyle \Delta H_{PZ} = \left(\frac{p_1}{\rho g}+z_1\right) - \left(\frac{p_2}{\rho g}+z_2\right)
$

which, with a horizontal meter $z_1 = z_2$, 

$
\displaystyle \Delta H_{PZ} = \frac{p_1}{\rho g}-\frac{p_2}{\rho g} = \frac{\Delta p}{\rho g}.
$

We have all of the data from the question now to calculate mass flow rate

$
\displaystyle \dot m = \rho \dot V = \rho c_dA_2\sqrt{\frac{2g\Delta p}{\rho g\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}= \rho c_dA_2\sqrt{\frac{2\Delta p}{\rho\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}
$

evaluating

$
\displaystyle \dot m = 1000\times0.96\times45\times10^{-4}\sqrt{\frac{2\times 12.36\times10^3}{1000\left(1-\left(\frac{45}{180}\right)^2\right)}}
$

$
\implies \underline{\dot m = 22.18\,\text{kg}/\text{s}.}
$

::::{admonition} Your work

You can now complete the {ref}`Fluid Motion chapter exercise page <section-fluid_motion_exercises>`.

You will revise Bernoulli and look at further flow measurement applications in the Spring Semester. For now - enjoy some thermodynamics!
