### Describing fluid flow

In general, fields of e.g. pressure, speed, temperature in a flow will be a function of both space, $x$, and time, $t$

$
p=p(x,t) \quad, \quad u=u(x,t) \quad, \quad T=T(x,t).
$

We can simplify to model flows in two classical ways. 

:::{hint} Concept: Steady flow
:class: concept

A steady flow does not change with time. Properties may vary from place to place, but are constant at any given place.

$
\implies p=p(x) \quad, \quad u=u(x) \quad, \quad T=T(x).
$

```{figure} images/steady_flow.png
:alt: Representation of steady flow in a flared pipe
:class: bg-none
:width: 80%
:align: center
```

The converse of steady flow is **unsteady flow**.

:::

:::{hint} Concept: Uniform flow
:class: concept

The properties in a uniform flow are the same at all points at any given instant, but they may change with time. 

$
\implies p=p(t) \quad, \quad u=u(t) \quad, \quad T=T(t).
$

```{figure} images/uniform_flow2.png
:alt: Representation of uniform flow in a pipe
:class: bg-none
:width: 80%
:align: center
```

The converse of uniform flow is **non-uniform flow**.

:::

:::{hint} Concept: 1-D flow
:class: concept

In 1-D flow, fluid properties are assumed to vary only along the flow direction, and to be uniform across any cross-section.

In practice, flow is rarely truly one-dimensional. For example, in a circular pipe, the velocity varies across the pipe, falling to zero at the wall and reaching a maximum towards the centre.

The 1-D assumption simplifies analysis and often provides sufficiently accurate results.

```{figure} images/1D_flow.png
:alt: Representation of 1-D flow in a pipe
:class: bg-none
:width: 80%
:align: center
```

:::

### Methods of highlighting flow pattern

A **streamline** a line parallel to the fluid flow at any instant in time. 

```{figure} images/streamlines.png
:alt: Streamlines over a kart
:class: bg-none
:width: 80%
:align: center
```

In uniform flow, streamlines will be **straight and parallel**. Streamlines give an instantaneous picture of the whole flow field.


A **streakline** joins, at a given time, all particles that have passed through a given point over a preceding period of time

```{figure} images/streaklines.jpg
:alt: Streaklines from an airofoil
:class: bg-none
:width: 50%
:align: center
```

VORTEX SHEDDING AT HIGH ANGLE OF ATTACKT. T. Lim, National University of Singapore, Singapore AIAA JOURNAL, Vol 36, no:10. 


A **pathline** shows the route taken by a single fluid particle during a given time interval.  It is equivalent to the trace of a single fluid particle seen on a long exposure photograph. Pathlines may cross.


In steady flow (not varying with time), pathlines, streaklines and streamlines all collapse onto the same trajectory.

<div style="display: flex; gap: 20px; justify-content: center; align-items: flex-start;">

  <div style="width: 30%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/stream_schem.png"
         style="max-width: 100%; max-height: 100%;"
         alt="Streamlines schematic">
  </div>

  <div style="width: 30%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/path_schem.png"
         style="max-width: 100%; max-height: 100%;"
         alt="Pathline schematic">
  </div>

  <div style="width: 40%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/steady_schem.png"
         style="max-width: 100%; max-height: 80%;"
         alt="Streamlines and pathlines the same in steady flow">
         </div>

</div>
