### Key assumptions

:::{hint}Concept: Ideal fluid
:class: concept

An **ideal fluid** is considered **inviscid** and {ref}`incompressible<concept_compressibility>`.

Inviscid: Viscosity is neglected, so the fluid offers no resistance to shear deformation.

Incompressible: Density is constant; changes in pressure do not produce significant changes in volume. This implies no phase change occurs and the fluid remains liquid or gas throughout.

We also assume that surface tension is negligible.

The converse of an ideal fluid is a **real** fluid.

:::

In general, fields of e.g. pressure, speed, temperature in a flow will be a function of both space, $x$, and time, $t$

$
p=p(x,t) \quad, \quad u=u(x,t) \quad, \quad T=T(x,t).
$

We can simplify to model flows in two classical ways. 

:::{hint} Concept: Steady flow
:class: concept

A steady flow does not change with time. Properties may vary from place to place, but are constant at any given place.

$
\implies p=p(x) \quad, \quad u=u(x) \quad, \quad T=T(x).
$

```{figure} images/steady.png
:alt: Representation of steady flow in a flared pipe
:class: bg-none
:width: 80%
:align: center
```

The converse of steady flow is **unsteady flow**.

:::

:::{hint} Concept: Uniform flow
:class: concept

The properties in a uniform flow are the same at all points at any given instant, but they may change with time. 

$
\implies p=p(t) \quad, \quad u=u(t) \quad, \quad T=T(t).
$

```{figure} images/uniform.png
:alt: Representation of uniform flow in a pipe
:class: bg-none
:width: 80%
:align: center
```

The converse of uniform flow is **non-uniform flow**.

:::

:::{hint} Concept: 1-D flow
:class: concept

In 1-D flow, fluid properties are assumed to vary only along the flow direction, and to be uniform across any cross-section.

In practice, flow is rarely truly one-dimensional. For example, in a circular pipe, the velocity varies across the pipe, falling to zero at the wall and reaching a maximum towards the centre.

The 1-D assumption simplifies analysis and often provides sufficiently accurate results.


:::

### Methods of highlighting flow pattern

Streamline

This is a line parallel to the fluid flow at any instant in time. 

```{figure} images/streamlines.png
:alt: Representation of 1-D flow in a pipe
:class: bg-none
:width: 80%
:align: center
```
