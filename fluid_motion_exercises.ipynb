### Viscosity

:::{exercise} 💧
A thin plate of dimensions $0.5\,\textsf{m}$ by $0.7\,\textsf{m}$ is supported horizontally by a thin oil film of thickness $0.5\,\textsf{mm}$.  If the oil has a viscosity of $0.29\,\textsf{kg}/\textsf{m}\,\textsf{s}$, estimate the force required to pull that plate at a constant velocity of $0.27\,\textsf{m}/\textsf{s}$, assuming a linear variation of velocity within the oil.

::::{dropdown} ✓ Check your answer

$54.81\,\textsf{N}$

::::

::::{admonition} Solution
:class: dropdown

Neglecting the resistance on the air side of the plate, we apply Newton's law of viscosity. 

$
\displaystyle \tau = \mu\frac{\partial u}{\partial y} \approx \mu \frac{u}{y}.
$

The force is the shear stress multiplied by the surface area applied

$
\displaystyle F=\tau A = \mu \frac{u}{y} A = 0.29 \frac{0.27}{0.5\times10^{-3}} 0.5\times0.7
$

$
\implies\underline{F=54.81\,\textsf{N}}.
$

The liquid space between two large flat parallel walls $25\,\text{mm}$ apart is filled with a liquid of viscosity $0.7\,\text{Pa}\,\text{s}$.  Within this space a thin, flat plate, $250\,\text{mm} \times 250\,\text{mm}$ is towed at a velocity of $150\,\text{mm}/\text{s}$ at a distance of $6\,\text{mm}$ from one wall, the plate and its movement being parallel to the walls.  Assuming linear variations of velocity between the plate and the walls, determine the force exerted by the liquid on the plate. 

::::{dropdown} ✓ Check your answer

::::

::::{admonition} Solution
:class: dropdown

```{figure} images/visc_qu.png
:alt: Shearing fluid with flat plate
:class: bg-none
:width: 80%

```

Newton's law of viscosity applies, and we note that the gap is small so the lubrication assumption can be made, with

$
\displaystyle \tau = \mu\frac{\partial u}{\partial y} = \mu\frac{u}{y},
$

with viscous force applied over the plate surface area $A = 0.25^2\,\text{m}^2$

$
\displaystyle F = \tau A = A\mu\frac{u}{y}.
$

On the top of the plate

$
\displaystyle F_t = 0.25^2\times0.7\frac{0.15}{0.006}
$

and the bottom surface

$
\displaystyle F_b = 0.25^2\times0.7\frac{0.15}{0.019}.
$

The total force on the plate is the sum of the forces on both surfaces

$
\displaystyle F = F_t + F_b = 0.25^2\times0.7\times 0.15\left(\frac{1}{0.006} + \frac{1}{0.019}\right)
$

$
\implies\underline{F = 1.44\,\text{N}.}
$



::::

:::{exercise} 📝
The space between two large flat parallel walls $25\,\textsf{mm}$ apart is filled with a fluid of viscosity $0.8\,\textsf{Pa}\,\textsf{s}$.  Within this space a thin flat plate, $200\,\textsf{mm}$ square, is pulled at $0.2\,\textsf{m}/\textsf{s}$ at a distance of $10\,\textsf{mm}$ from one wall in a direction parallel to the side walls.  Assuming linear variations of velocity between the plate and the walls, determine the force required to pull the plate.  

::::{dropdown} ✓ Check your answer

$1.07\,\textsf{N}$

::::

::::{admonition} Solution
:class: dropdown

```{figure} images/visc_q2.png
:alt: Shearing plates
:class: bg-none
:width: 80%
:align: center
```

:::{exercise} 📝
A hydraulic ram of diameter $d=80\,\textsf{mm}$ and length $l=100\,\textsf{mm}$ moves wholly within a concentric cylinder $80.2\,\textsf{mm}$ in diameter. The annular clearance is filled with oil of relative density $0.7$ and kinematic viscosity $410\,\textsf{mm}^2/\textsf{s}$.  

1. What is the viscous force resisting the motion when the ram moves at $105\,\textsf{mm}/\textsf{s}$?
    
2. Assuming the ram moves continuously backwards and forwards at constant speed (ie neglecting acceleration and deceleration at the end of each stroke) how much power is dissipated?


::::{dropdown} ✓ Check your answer

1. $7.6\,\textsf{N}$
2. $0.8\,\textsf{W}$

::::

::::{admonition} Solution
:class: dropdown

1. Newton's law of viscosity gives us

$
\displaystyle \tau \approx \mu\frac{u}{y}.
$

Noting that the kinematic viscosity $\nu = \mu/\rho$, $\mu = \nu \rho$, we have

$
\displaystyle \tau = \nu \rho\frac{u}{y}.
$

The force applied $F=\tau A$ where the area the shear stress is applied over is the cylindrical surface of the ram. Therefore, for the target speed $u = 105\,\textsf{mm}/\textsf{s}$

$
\displaystyle F = \nu \rho\frac{u}{y} \pi dl = 410 \times 10^{-6}\times 700\frac{0.105}{0.1\times10^{-3}} \pi 0.08\times0.1
$

$
\implies\underline{F = 7.6\,\textsf{N}}.
$

2. Power = force $\times$ velocity

$
P = Fu = 7.6\times 0.105
$

$
\implies\underline{P=0.8\,\textsf{W}}.
$

:::{exercise} 📝
A simple journal bearing of internal diameter $40.16\,\textsf{mm}$ and length $30\,\textsf{mm}$ supports a shaft of diameter $40\,\textsf{mm}$ turning at $600\,\textsf{rpm}$. The gap between shaft and bearing is completely filled with oil of viscosity $0.17\,\textsf{kg}/\textsf{m}\,\textsf{s}$.  

1. Approximately how much power is dissipated in the bearing?

2. Give two reasons why this answer is only approximate.

::::{dropdown} ✓ Check your answer

$12.6\,\textsf{W}$

::::

::::{admonition} Solution
:class: dropdown

1. Newton's law of viscosity

$
\displaystyle \tau = \mu\frac{\partial u}{\partial y} \approx \mu \frac{u}{y}.
$

The speed of the shaft $u$ is related to the rotation rate $u = r\omega$, where $r$ is the shaft radius. Therefore 

$
\displaystyle \tau \approx \mu \frac{r\omega}{y}
$

giving the force

$
\displaystyle F = \tau A = A \mu \frac{r\omega}{y}
$

and power (noting that $\omega_{\text{rad/s}} = 2\pi \omega_{\text{RPM}}/60$)

$
\displaystyle P = Fu = \tau A r \omega = A \mu \frac{(r\omega)^2}{y} = (\pi 0.04\times 0.03)\times 0.17\frac{0.02^2}{0.08\times 10^{-3}}\left(600\frac{2\pi}{60}\right)^2
$ 

$
\implies \underline{P=12.6\,\textsf{W}}.
$

2. 
- The linear velocity gradient is only an approximation.
- The shaft will not run at the exact centre of the bearing and therefore the gap width varies around the shaft
- Viscosity is a function of temperature and will therefore vary until steady conditions are obtained

### Mass conservation - Continuity

:::{exercise}💧

Water of density $1000\,\text{kg}/\text{m}^3$ and viscosity $0.001\,\text{Pa}\,\text{s}$ flows steadily along a pipe of internal diameter $3\,\text{mm}$.  $200\,\text{ml}$ of water are collected in a measuring beaker in $43\,\text{s}$.  

1. What is the Reynolds number of the pipe flow?  
2. Is this flow laminar, transitional or turbulent?



```{figure} images/we14.png
:alt: Water flowing into a measuring cylinder
:class: bg-none
:width: 40%
:align: center
```

::::{dropdown} ✓ Check your answer

1. $1980$
2. The pipe flow is laminar, but on the verge of transition. 

::::

::::{admonition} Solution
:class: dropdown

1. To calculate the Reynolds number we need to find the flow speed in the pipe. For a control volume intersection pipe and beaker, the volume flow rate in (from the pipe) equals the rate of accumulation of water in the beaker. There are no volme fluxes out. Therefore

$
\displaystyle Q_p = \frac{\Delta V}{\Delta t} = u\frac{\pi d^2}{4},
$

which, rearranging for $u$, is

$
\displaystyle u = \frac{4\Delta V}{\pi d^2 \Delta t}= \frac{4\times 200 \times 10^{-6}}{\pi (3\times 10^{-3})^2 \times 43}
$

$
\implies u = 0.66\,\textsf{m}/\text{s}.
$

Therefore the Reynolds number in the pipe flow

$
\displaystyle \text{Re} = \frac{\rho u d}{\mu} = \frac{1000\times 0.66 \times 3\times10^{-3}}{0.001}
$

$
\implies\underline{\text{Re}=1980.}
$

2. This Reynolds number is slightly below the threshold value for pipe flow of $2000$. Therefore the flow is laminar but very close to the point of transition. 

::::

:::{exercise}💧

An ideal fluid flows along the pipe shown.  If the inlet velocity is $25\,\text{m}/\text{s}$, the inlet diameter $12\,\text{mm}$ and the outlet diameter $28\,\text{mm}$, what is the outlet velocity?

```{figure} images/cons_ex.png
:alt: Pipe with an expansion
:class: bg-none
:width: 80%
:align: center
```

```{figure} images/cons_ex_cv.png
:alt: Pipe with an expansion - showing control volume
:class: bg-none
:width: 80%
:align: center
```

For an ideal fluid, which is incompressible, $\rho$ is constant in the pipe and mass conservation reduces to continuity. For the control volume shown with steady flow

$
\displaystyle \dot V_{\text{in}} = \dot V_{\text{out}} \quad \implies \quad u_{\text{in}}A_{\text{in}} = u_{\text{out}}A_{\text{out}} \quad \implies \quad u_{\text{in}}\frac{\pi d_{\text{in}}^2}{4} = u_{\text{out}}\frac{\pi d_{\text{out}}^2}{4}.
$

Therefore the outlet velocity

$
\displaystyle u_{\text{out}} = u_{\text{in}}\frac{d_{\text{in}}^2}{d_{\text{out}}^2} = 25\frac{12^2}{28^2}
$

$
\implies\underline{u_{\text{out}} = 4.59\,\text{m}/\text{s}.}
$

::::

(exercise_duct_cylinder)=
:::{exercise}📝
Water flows along a rectangular duct with cross-section $0.3\,\textsf{m}\times 0.25\,\textsf{m}$ with a mean velocity of $0.5\,\textsf{m}/\textsf{s}$.  There is a cylinder within the duct with a diameter of $0.2\,\textsf{m}$, as shown in Figure Q8.  

Calculate:

1. the mass flow rate of the water along the duct.
2. the velocity ($u_2$) of the water in the space between the cylinder and the duct.

```{figure} images/fm_q8.png
:alt: Rectangular duct with cylinder
:class: bg-none
:width: 80%
:align: center
```

1. Calculate the volume flow rate.
2. The pipe contracts to a diameter (2) of $28\,\textsf{mm}$.  Find the velocity and pressure in this section of pipe assuming that the density is constant throughout.

Figure Q13: Pipe network diagram, for use with exercise 13 (not to scale).


::::{dropdown} ✓ Check your answer

1. $3.12\times 10^{-3}\,\textsf{m}^3/\textsf{s}$
2. $5.06\textsf{m}/\textsf{s}$, $1.01\,\textsf{bar}$
3. $5.06\textsf{m}/\textsf{s}$, $72\,\textsf{kPa}$

::::

::::{admonition} Solution
:class: dropdown

1. Volume flow rate (by definition) is the flow speed multiplied by the cross-sectional area through which it passes

$
\displaystyle \dot V = uA = u\left(\frac{\pi d^2}{4}\right).
$

Evaluating

$
\displaystyle \dot V = 2.9\left(\frac{\pi 0.037^2}{4}\right)
$

$
\implies\underline{\dot V = 3.12\times 10^{-3}\,\textsf{m}^3/\textsf{s}.}
$

2. The velocity follows from applying mass conservation to the control volume intesecting 1 and 2 normal to the flow direction

$
\dot m = \rho_1A_1u_1 = \rho_2A_2u_2
$

which with constant density reduces to continuity and

$
\displaystyle u_2 = \left(\frac{A_1}{A_2}\right)u_1 =  \left(\frac{d_1^2}{d_2^2}\right)u_1.
$

Applying Bernoulli along the horizontal streamline connecting 1 and 2

$
\displaystyle p_1 + \frac{\rho u_1^2}{2} = p_2+ \frac{\rho u_2^2}{2},
$

which rearranges, and combines with continuity

$
\displaystyle p_2 = p_1 + \frac{\rho (u_1^2-u_2^2)}{2} = p_1 + \frac{\rho u_1^2(1-d_1^4/d_2^4)}{2}.
$

Evaluating,

$
\displaystyle p_2 = 1.1\times10^5 + \frac{1000\times2.9^2(1-0.037^4/0.028^4)}{2}
$

$
\implies \underline{p_2 = 101 403\,\textsf{Pa} = 1.01\,\textsf{bar}.}
$

3. From 2 to 3 the diameter of the pipe is unchanged and the density of the water is unchanged therefore, by continuity, the flow velocity is unchanged,

$
\implies \underline{u_3 = 5.06\,\textsf{m}/\textsf{s}.}
$

Applying Bernoulli along the streamline connecting 2 to 3

$
\displaystyle  p_2+\rho g z_2 + \frac{\rho u_2^2}{2} = p_3+\rho g z_3 + \frac{\rho u_3^2}{2}.
$

With $u_2 = u_3$ the dynamic pressure terms cancel, and

$
p_3 = p_2 +\rho g(z_2-z_3) = 1.01\times10^5 + 1000\times9.81(-3)
$

$
\implies\underline{p_3 = 71973\,\textsf{Pa} = 72.0\,\textsf{kPa}.}
$

::::

:::{exercise}📝

For a pipe carrying water in a horizontal plane branching with the geometry shown in Figure Q14, find pressures $p_2$ and $p_3$ (gauge). 

```{figure} images/fm_q14.png
:alt: Branching pipe, horizontal plane
:class: bg-none
:width: 60%
:align: center
```

::::{dropdown} ✓ Check your answer

$13.17\,\textsf{kPa}$

::::

::::{admonition} Solution
:class: dropdown

The mass flow rate is given by the venturimeter rate equation

$
\displaystyle \dot m_{\text{real}} = c_d\rho\dot V_{\text{real}} = c_d\rho A_2\sqrt{\frac{2g\Delta H_{PZ}}{\left(1-\left(\frac{A_2}{A_1}\right)^2\right)}}
$

which we can rearrange to find the piezometric head and hence the pressure. Rearranging

$
\displaystyle \Delta H_{PZ} = \left(\frac{\dot m_{\text{real}}}{c_d\rho A_2}\right)^2\left(\frac{1-\left(\frac{A_2}{A_1}\right)^2}{2g}\right)
$

or writing areas in terms of diameters

$
\displaystyle \Delta H_{PZ} = \left(\frac{\dot m_{\text{real}}}{c_d\rho \pi d_2^2/4}\right)^2\left(\frac{1-\left(\frac{d_2}{d_1}\right)^4}{2g}\right) = \left(\frac{60}{0.97\times750\pi 0.15^2/4}\right)^2\left(\frac{1-\left(\frac{0.15}{0.3}\right)^4}{2\times9.81}\right)
$

$
\implies \Delta H_{PZ} = 1.04\,\textsf{m}.
$

But the piezometric head relates to pressures

$
\Delta H_{PZ} = \frac{p_1-p_2}{\rho g} + (z_1-z_2).
$

Therefore the pressure difference

$
p_1 - p_2 = \rho g\left(\Delta H_{PZ} - (z_1-z_2)\right) = 750\times9.81(1.04 - (-0.75))
$

$
\implies \underline{p_1 - p_2 = 13.17\,\textsf{kPa}.}
$

::::

::::{exercise}🔥

Consider the vertical venturimeter, shown in Figure Q16. 

```{figure} images/fm_q16.png
:alt: Vertical venturimeter and manometer
:class: bg-none
:width: 75%
:align: center
```

2. Use the information provided to find the volume flow rate through the venturimeter. See if you can do this without looking up the equation, but using the manometer measurement and principles of Bernoulli.

:::{dropdown} ✓ Check your answer

2. $1.3\times10^{-3}\,\text{m}^3/\text{s}$

::::

::::{admonition} Solution
:class: dropdown

  Let $X$ denote the interface level in the left-hand limb of the manometer. Let $H=600\,\text{mm}$ denote the height difference between points 1 and 2, and let $L=430\,\text{mm}$ denote the height difference between the interfaces in the right- and left-hand limbs of the manometer. Let $h_1$ denote the height between point 1 and $X$, and $h_2$ denote the height between point 2 and the interface in the right-hand limb of the manometer. Hence, $H = h_1 + L + h_2$.

  Hence, from the manometer we have

$
p_X = p_1 - \rho_o g h_1\quad(\text{left-hand limb}),
$

$
p_X = p_2 + \rho_o g h_2 + \rho_m g L\quad(\text{right-hand limb}).
$

Eliminating $p_X$ gives

$
p_1 - p_2 = \rho_o g(h_2 + h_1) + \rho_mgL= \rho_o g(H-L) + \rho_mgL=\rho_o gH + (\rho_m - \rho_o)gL,
$

where in the above we have used $H = h_1 + L + h_2$.

Applying Bernoulli's equation along a streamline connecting points 1 and 2 gives

$
\displaystyle \frac{p_1}{\rho_o g} + \frac{u_1^2}{2g} = \frac{p_2}{\rho_o g} + \frac{u_2^2}{2g} + H
$

giving
$
\displaystyle p_1 - p_2 = {\textstyle\frac{1}{2}}\rho_0\big(u_2^2 - u_1^2\big) + \rho_0 g H
$

and

$
\displaystyle p_1 - p_2 = {\textstyle\frac{1}{2}}\rho_0\left(\frac{\dot V^2}{A_2^2} - \frac{\dot V^2}{A_1^2}\right) + \rho_0 g H,
$

where $\dot V=A_1u_1=A_2u_2$. 

Eliminating $p_1 - p_2$ from manometer and from Bernoulli gives

$
\displaystyle  \rho_o gH + (\rho_m - \rho_o)gL = {\textstyle\frac{1}{2}}\rho_0\left(\frac{A_1^2 - A_2^2}{A_1^2A_2^2}\right)\dot V^2+ \rho_0 g H
$

$
\displaystyle \dot V = A_1A_2\sqrt{\frac{2gL}{A_1^2 - A_2^2}\left(\frac{\rho_m}{\rho_o}-1\right)}= {\textstyle\frac{1}{4}}\pi d_1^2d_2^2\sqrt{\frac{2 g L}{d_1^4 - d_2^4}\left(\frac{\rho_m}{\rho_o} - 1\right)}.
$

Note, this is the same as our standard equation for a venturimeter - that is, the angle of inclination makes no difference. 

Substituting in $d_1=0.105\,\text{m}$, $d_2=0.035\,\text{m}$, $L=0.43\,\text{m}$, $\rho_m/\rho_0=1000/830=1.2$ 

$
\implies\underline{\dot V = 1.3\times10^{-3}\,\text{m}^3/\text{s}.}
$
