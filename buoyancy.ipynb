(section-buoyancy)=
# Buoyancy



:::{dropdown} What happens when a body is immersed in a fluid?

The pressure surrounding the body increases with depth

$
p=p_0+\rho gh.
$

:::{dropdown} What is the result of that pressure change?

Pressure forces act normal to the surface therefore 

- horizontal components cancel

- vertical components **do not** cancel

and there is resultant upward force on the body. This force is **buoyancy**.

:::{dropdown} What is the magnitude of the buoyancy force?

Imagine replacing the body with the exact volume of fluid it displaced. That fluid would be in equilibrium, so **the buoyancy force is equal to the weight of the displaced fluid**.

:::


:::{tip} Definition: Archimedes Principle
:class: definition

A wholly or partially immersed body experiences an upward buoyancy force equal to the weight of the fluid it displaces

$
F_B = \rho_{\text{fluid}}V_{\text{displaced}}g.
$

:::


An immersed body experiences two vertical forces, its weight

$
W=\rho_{\text{body}} g V
$

and buoyancy

$
F_B=\rho_{\text{fluid}} g V,
$

The behaviour of the body depends on their relative magnitudes.

What makes a body rise, sink or float?

:::{dropdown} ↑ Rising

$
\implies F_B>W
$

The resultant force is upward, so the body accelerates upwards.

As it rises, less of the body may be immersed → $V_{\text{displaced}}$ decreases → $F_B$ decreases.

Eventually

$
F_B=W
$

and the body floats.

:::

:::{dropdown} ↓ Sinking

$
\implies F_B<W
$

The resultant force is downward, so the body accelerates downwards.

If it remains fully immersed, $V_{\text{displaced}}=V$, so the buoyancy force remains constant.

:::

:::{dropdown} ↔ Floating

$ 
\implies F_B=W
$

The body is at equilibrium. For a floating body,

$
\rho_f gV_{\text{displaced}} = \rho gV_{\text{body}} 
$

so

$
\displaystyle \frac{V_{\text{displaced}}}{V_{\text{body}}} =\frac{\rho}{\rho_f}
$

A body whose average density is less than that of the surrounding fluid can float with only part of its volume submerged.

:::


:::{hint} Concept: 💭 Thought experiment- Brick in a boat
:class: concept

You are sitting in a boat floating on a lake.

You have a brick in the boat.

You throw the brick into the water.

What happens to the water level? Does it 

A. Rise $\quad$ B. Fall $\quad$ or C. Stay the same?

:::{dropdown} See explanation

**Before:** the boat supports the weight of the brick with

$
F_{B1} = W_{\text{boat}} + W_{\text{brick}} = V_{\text{disp1}}\rho_w g,
$

and the brick's weight determines how much water the boat displaces.

**After:** the boat only supports its own weight, so it rises relative to the water with

$
F_{B2} = W_{\text{boat}} = V_{\text{disp2}}\rho_w g.
$

The change in water volume displaced by the boat is therefore 

$
V_{\text{disp1}} - V_{\text{disp2}}=m_{\text{brick}}/\rho_w,
$

and $V_{\text{disp2}}$ must be less than $V_{\text{disp1}}$. 

But the brick also sinks and displaces water equal to its own volume $V_{\text{brick}} = m_{\text{brick}}/\rho_b$, leading to a rise in the water level.

For a brick, which is denser than water, the volume displaced by the submersed brick is smaller than the change in water displaced by the removal of the brick from the boat

$
m_{\text{brick}}/\rho_b < m_{\text{brick}}/\rho_w.
$

**Answer B. The water level falls**. 

:::{dropdown} 🪵 What if it were a block of wood?

Now replace the brick with a block of wood that floats.

What happens when you throw it overboard?

:::{dropdown} See explanation

In the boat, the wood makes the boat displace water with a weight equal to the weight of the wood.

Once the wood is floating separately, it displaces water with a weight equal to its own weight.

So the total displaced water **remains unchanged**.

::::

:::{tip} Definition: Centre of buoyancy
:class: definition

The buoyancy force acts through the **centre of gravity of the volume of fluid displaced**, termed the **centre of buoyancy**. 

The centre of buoyancy does not in general correspond to the centre of gravity of the body. If the fluid density is constant, the centre of gravity of the displaced fluid is at the centroid of the immersed volume.

::: 

:::{hint} Concept: 💭 Thought experiment- What if the centre of buoyancy is displaced horizontally?
:class: concept

The weight $W$ acts downward through the centre of gravity, while $F_B$ acts upward through the centre of buoyancy.

```{figure} images/tipping.png
:alt: Representation of buoyant stability
:class: bg-none
:width: 95%
:align: center
```

The relative positions of the centre of gravity and centre of buoyancy therefore determine whether a floating body tends to return to its original orientation or overturn. This is the concept of **buoyant stability**.

::::{prf:example} Floating cube
:enumerator: 9

A cube of side $l=35\,\textsf{mm}$ is floating in water of density $\rho_w = 1000\,\textsf{kg}/\textsf{m}^3$.

1. What is the density of the cube material if the depth of submersion $h_i = 27\,\textsf{mm}$?

2. A hemispherical scoop, diameter $d =15\,\textsf{mm}$, is removed from the top face. What is the new depth of immersion?

:::{hint}
:class: dropdown

The volume of a sphere with radius $r$, diameter $d=2r$ is 

$
\displaystyle V=\frac{4}{3}\pi r^3 = \frac{\pi d^3}{6}.
$

:::

:::{admonition} Solution
:class: dropdown

1.  

```{figure} images/we9a.png
:alt: Floating cube showing depth of immersion
:class: bg-none
:width: 60%
:align: center
```
The floating cube is in equilibrium, therefore it's weight balances the upward buoyancy force, which is the weight of the displaced fluid volume

$
F_B=h_il^2\rho_w g = \rho l^3 g.
$

Rearranging for the cube density $\rho$

$
\displaystyle \rho = \frac{h_i\rho_w}{l} = \frac{27\times 1000}{35}\,\frac{\textsf{mm}\,\textsf{kg}}{\textsf{mm}\,\textsf{m}^3}
$

$
\implies \underline{\rho = 771\,\textsf{kg}/\textsf{m}^3}.
$



2. 

```{figure} images/we9b.png
:alt: New depth of immersion for floating cube with scoop removed
:class: bg-none
:width: 60%
:align: center
```

Now the weight of the cube is changed, because some material is removed. The new equilibrium is given by 

$
\displaystyle F_B=hl^2\rho_w g = \rho g \left(l^3 - \frac{1}{2}\frac{\pi d^3}{6}\right).
$

Rearranging for the new immersion depth 

$
\displaystyle h = \frac{\rho}{l^2\rho_w} \left(l^3 -\frac{\pi d^3}{12}\right) = \frac{771}{35^2\times 1000} \left(35^3 -\frac{\pi 15^3}{12}\right)
$


$
\implies \underline{h = 26.4\,\textsf{mm}}.
$

::::{prf:example} Helium balloon
:enumerator: 10

A helium balloon has a mass of $300\,\textsf{kg}$ (balloonist, balloon structure etc) with an additional $400\,\textsf{kg}$ of sand ballast is also carried.

```{figure} images/met_balloon_qu.png
:alt: Meteorological balloon being filled with helium gas
:class: bg-none
:width: 60%
:align: center
```

1. At ground level the air pressure is $1.01\,\textsf{bar}$ and the temperature is $300\,\textsf{K}$. Calculate the volume of helium that must be put into the balloon so that it can rise.

($\tilde R$ for air is $287.1\,\textsf{J}/\textsf{kg}\,\textsf{K}$; $\tilde R$ for helium is $2077\,\textsf{J}/\textsf{kg}\,\textsf{K}$).

2. The balloon rises to $2000\,\textsf{m}$ where the air pressure is $0.795\,\textsf{bar}$ and the temperature is $275\,\textsf{K}$. To what volume will the balloon expand?

3. The balloon is now completely full and cannot expand any more. Helium must be vented if it rises any higher to prevent it bursting. The balloon needs to rise further to get over some mountains. How much sand ballast must be released if the balloon is to rise to $3000\,\textsf{m}$ where the air pressure is $0.7012\,\textsf{bar}$ and the temperature is $269\,\textsf{K}$?

4. Show that the buoyancy of the balloon is related only to the mass of helium it is filled with, provided that the balloon is in equilibrium with the surrounding air.

:::{admonition} Solution
:class: dropdown

1. The total mass of the balloon plus ballast plus additional mass of helium added must be balanced by the buoyancy of the added helium which is the weight of air displaced

$
m_bg+V_b\rho_{He}g = V_b\rho_{\text{air}}g,
$

$
m_b = V_b(\rho_{\text{air}}-\rho_{He}).
$

From the Perfect Gas Law, we can relate the densities to the gas constants provided in the question, since 

$
\displaystyle p=\rho\tilde R T \quad\implies\quad \rho =\frac{p}{\tilde R T}.
$

Therefore 

$
\displaystyle m_b=\frac{pV_b}{T}\left(\frac{1}{\tilde R_{\text{air}}}-\frac{1}{\tilde R_{He}}\right)\quad\dots(\dagger)
$

and

$
\displaystyle V_b=\frac{Tm_b}{p}\left(\frac{1}{\tilde R_{\text{air}}}-\frac{1}{\tilde R_{He}}\right)^{-1}\quad\dots(\ast)
$


At ground level $p=1.01\,\textsf{bar}$ and the temperature is $300\,\textsf{K}$; evaluating for the volume of helium to be added at ground level for the balloon to rise

$
\displaystyle V_b=\frac{300\times 700}{1.01\times 10^5}\left(\frac{1}{287.1}-\frac{1}{2077}\right)^{-1}
$

$
\implies\underline{V_b=692\,\textsf{m}^3}.
$

2. The balloon rises to $2000\,\textsf{m}$ where the air pressure is $0.795\,\textsf{bar}$ and temperature $275\,\textsf{K}$ (assuming the balloon is in equilibrium with the air). From equation ($\ast$)

$
\displaystyle V_b=\frac{275\times 700}{0.795\times 10^5}\left(\frac{1}{287.1}-\frac{1}{2077}\right)^{-1}
$

$
\implies\underline{V_b=806.6\,\textsf{m}^3},
$

at $2000\,\textsf{m}$.

3. The balloon rises further, to $3000\,\textsf{m}$, where the air pressure is $0.7012\,\textsf{bar}$ and temperature $269\,\textsf{K}$, but the volume remains at $V_b=806.6\,\textsf{m}^3$. Sand ballast must be ditched. From ($\dagger$), the target mass of the balloon will be 

$
m_b = \frac{0.7012\times10^5\times 806.6}{269}\left(\frac{1}{287.1}-\frac{1}{2077}\right)
$

$
\implies m_b=631.2\,\textsf{kg}.
$

So the balloonist must jettison $\underline{700-631.2=68.8\,\textsf{kg}}$ of sand ballast. 

4. The Perfect Gas Law for helium states

$
\displaystyle \frac{pV_b}{T} = m_{He}\tilde R_{He}
$

(the helium has displaced any air in the balloon). Combining with equation ($\dagger$)

$
\displaystyle m_b=\frac{pV_b}{T}\left(\frac{1}{\tilde R_{\text{air}}}-\frac{1}{\tilde R_{He}}\right)
$

$
\displaystyle m_b=m_{He}\left(\frac{\tilde R_{He}}{\tilde R_{\text{air}}}-1\right).
$

Therefore the mass that can be lifted $m_b$ depends only on the mass of helium added.
