# Density

**Density** is a concentration of mass $\Leftrightarrow$ a mass per unit volume.

With the standard symbols:

| Symbol | Quantity | SI unit |
|:---:|:---|:---:|
| $\rho$ | Density | $\mathsf{kg/m^3}$ |
| $m$ | Mass | $\mathsf{kg}$ |
| $V$ | Volume | $\mathsf{m}^3$ |


:::{tip} Definition: Density
:class: definition
The density of a fluid is its mass per unit volume,

$
\displaystyle\rho = \frac{m}{V}.
$
:::

It is always a good plan to stick with SI units. This makes it easier to spot mistakes and work out units you are not sure about. 

:::{prf:example} Density
Calculate the density of $1.2\,\mathsf{m}^3$ water with mass $1200\,\mathsf{kg}$.

::::{admonition} Solution
:class: dropdown
To find the density we need the volume of a known mass of fluid, or the mass of a known volume of fluid. 

A fluid density $\rho=m/V$, where $m$ is the mass and $V$ the volume of the fluid respectively. Here, the water has mass
   $m=1200\,\mathsf{kg}$ and volume $V=1.2\,\mathsf{m}^3$. Hence,
$$
\rho=\frac{1200}{1.2}\,\mathsf{kg}/\mathsf{m}^3
$$
$$
\implies \underline{\rho = 1000\,\mathsf{kg}/\mathsf{m}^3}.
$$
::::

### Alternative definitions of mass concentration

| Symbol | Quantity | SI unit |
|:---:|:---|:---:|
| $\Rho$ | Relative density |  |
| $\mathcal{v}$ | Specific volume | $\mathsf{m}^3/\mathsf{kg}$ |

Sometimes you see the density of a fluid described relative to the density of water. This is the **relative density**; the ratio of a fluid density to the density of water, with ($\rho_w=1000\,\mathsf{kg/m}^3$).

An alternative formulation you often see is the inverse of density, or specific volume. The word **specific** means "per unit mass", so specific volume is the "volume per unit mass", which is $1/\rho$.

:::{tip} Definition: Specific volume
:class: definition
The specific volume of a fluid is its volume per unit mass; the inverse of its density,

$
\displaystyle \mathcal{v} =\frac{V}{m}= \frac{1}{\rho}.
$
:::

:::{prf:example} Specific volume
A sample of water has a mass of $4.0\,\mathsf{kg}$ and occupies a volume of $0.0040\,\mathsf{m}^3$.
Calculate the specific volume of the water.

::::{admonition} Solution
:class: dropdown

The specific volume of the water $\mathcal{v}=V/m$ ($=1/\rho$) with mass $m=4.0\,\mathsf{kg}$ and volume $V=0.0040\,\mathsf{m}^3$ respectively. Hence, 
$$
  \mathcal{v}=\frac{0.0040}{4}\,\mathsf{m}^3/\mathsf{kg}
   $$
$$
\implies \underline{\mathcal{v} = 0.001\,\mathsf{m}^3/\mathsf{kg}}.
$$

::::

Bulk density refers to the average density of an inhomogeneous substance i.e.~the total mass per unit of mixture volume. 

:::{prf:example} Bulk density
A tank contains a water–solid suspension with a total mass of $120\,\mathsf{kg}$. The suspension occupies $0.100\,\mathsf{m}^3$.
Calculate the bulk density of the suspension.

::::{admonition} Solution
:class: dropdown

The bulk density of the suspension $\rho=m/V$, where the total mass $m=120\,\mathsf{kg}$ and total volume $V=0.100\,\mathsf{m}^3$. Hence,

   $$
   \rho=\frac{120}{0.100}\,\mathsf{kg}/\mathsf{m}^3
   $$   

   $$
   \implies \underline{\rho = 1200\,\mathsf{kg}/\mathsf{m}^3}.
   $$
::::

### Key assumptions

Fluids and especially gases comprise atoms and molecules with much space between (and within) them. We ignore these molecular level variations in mass distribution and consider most engineering fluid flows to be **homogeneous**. 

:::{hint} Concept: Homogeneity
:class: concept

A homogeneous fluid is one for which the density is assumed to be uniform throughout the fluid, so that a single density value can be used to represent the fluid.
