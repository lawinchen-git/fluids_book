# Fluid statics

### Learning outcomes
- Understand the origins of buoyancy in fluids
- Calculate buoyancy forces and equibilbrium positions of fully or partially submerged objects
- Understand how pressure acts on submerged surfaces
- Develop and apply equations to calculate hydrostatic forces on submerged surfaces

This chapter uses the concepts of {ref}`density <section-density>` and {ref}`pressure <section-pressure>` you have learned about in "Fluid Properties" to understand how stationary fluids behave. We will see how the hydrostatic pressure variation in a static fluid is critical to it's interaction with bodies and surfaces.
