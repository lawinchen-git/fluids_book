# Pressure

### Pressure in a fluid
In a fluid (liquid or gas) molecules are in continuous motion $\rightarrow$ There is an amount of energy associated with this motion.

```{figure} images/pressure_schem.png
:alt: Representation of pressure through molecular motion
:class: bg-none
:width: 80%
:align: center
```

When the molecules interact with each other there is an equal and opposite reaction - therefore no net force. But when the molecules interact with walls, there is a resultant force perpendicular to the surface. Each impact force is tiny, but there are **many** of them. 

<div style="display: flex; gap: 20px; justify-content: center; align-items: flex-start;">

  <div style="width: 45%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/mol_interact.png"
         style="max-width: 100%; max-height: 100%;"
         alt="Molecular interactions with each other">
  </div>

  <div style="width: 45%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/mol_wall.png"
         style="max-width: 100%; max-height: 80%;"
         alt="Molecular interactions with a wall">
  </div>

</div>

If we consider a tiny element of wall area $\delta A$, we can imagine a single molecule exerting a force $\delta F$ on that. The pressure exerted on the wall by the molecule is defined as the force per unit area, or

$\displaystyle p=\frac{\delta F}{\delta A}$. 

```{figure} images/pressure_force.png
:alt: Representation of pressure through molecular motion
:class: bg-none
:width: 80%
:align: center
```

To find the total force, and hence pressure, we need to sum up all the molecular-level contributions over the full surface, i.e.~$F = \sum \delta F$. 

In the limit $\delta A,\,\,\delta F \rightarrow 0$, $F \rightarrow \int\,dF$, and

$F=\int p\,dA$.

In the case of constant pressure, we can integrate with

$\displaystyle F = \frac{p}{A}\quad, \quad p=\frac{F}{A}$.

These definitions hold if the pressure is constant or an average value is appropriate. In many instances this is not the case - to be discussed!

:::{tip} Definition: Pressure
:class: definition
Pressure is an energy density within a fluid, characterised by molecular-level motion. This equates to the force ($F$) exerted by a fluid per unit area ($A$) of surface to which it is applied.

$
\displaystyle p=\frac{F}{A}.
$
:::

**Notes about pressure**
- Pressure is a **scalar** quantity, acting equally in all directions. Compared to a force, which is a vector with magnitude and direction.
- Pressure has the units of force per unit area, which is a **stress**.
- Gases cannot sustain tensile stress, liquids a negligible amount.
- The force exerted on a vessel by fluid pressure acts perpendicular (normal) to the vessel surface. 
- The vessel walls exert an equal and opposite force on the fluid.


### Pressure and temperature in gases

Molecular fluctuations are also responsible for heat (as indicated by temperature), meaning there is a relationship between the pressure and temperature of a fluid. Hence why you study fluid mechanics and thermodynamics in the same module!

:::{hint} Concept: A Perfect Gas
:class: concept
In a **Perfect Gas** the molecules behave as perfectly elastic small spheres, influencing each other only when they collide and so smalle that they occupy a tiny proportion of the volume of space in which they move.
:::

No real gas is a perfect gas, though many behave nearly so. Luckily, air is close to a perfect gas. 

In a perfect gas, there is a direct relationship between pressure and temperature. 

| Symbol | Quantity |  | SI unit |
|:---:|:---|:---:|:---:|
| $p$ | Pressure |  | $\mathsf{Pa}$ |
| $T$ | Temperature | | $\mathsf{K}$ |
| $R$ | Universal Gas Constant  | 8.31445 | $\mathsf{J/mol\,K}$ |
| $\mathcal{M}$ | Molar mass | | $\mathsf{kg/mol}$ |
| $n = m/\mathcal{M}$ | Number of moles |  | $\mathsf{mol}$|
| $\tilde R = R/\mathcal{M}$ | Specific Gas Constant  | | $\mathsf{J/kg\,K}$ |

A common unit of pressure (not SI) is $\mathsf{bar}$, where $1\,\mathsf{bar}=10^5\mathsf{Pa}$. This is convenient, because $1\,\mathsf{bar}$ is close to atmospheric pressure. 

:::{tip} Definition: The Perfect Gas Law
:class: definition

The product of the pressure and volume of a perfect gas is directly proportional to the product of its absolute temperature and amount of substance, with a universal constant of proportionality $R$ - the Universal Gas Constant. Hence,

$
\displaystyle pV=nRT \quad\mathrm{or}\quad p=\rho RT,
$

where $R=8.31445\,\mathsf{J/mol\,K}$.
:::


:::{prf:example} Perfect gas law 
:enumerator: 4
A hydrogen balloon is to expand to a sphere of $20\,\mathsf{m}$ diameter at a height of $30\,\mathsf{km}$ where the air pressure is $1100\,\mathsf{Pa}$ and the temperature is $-40^\circ\mathsf{C}$. If there is to be no stress in the fabric of the balloon (i.e.~the pressure inside the balloon equals the pressure outside), what volume of hydrogen must be added at ground level where the air pressure is $101.3\,\mathsf{kPa}$ and the temperature is $15^\circ\mathsf{C}$?

::::{admonition} Solution
:class: dropdown
So we have 2 different states we need to look at; at ground level (1) and at $30\,\mathsf{km}$ (2). We know the temperature and pressure at each state and we know that the hydrogen obeys the Perfect Gas law. The conditions at the two elevations give us the proportionality in the law, which relates to the amount of gas and therefore the volume we are asked for. 

At ground level $p_1V_1=nRT_1$, and at $30\,\mathsf{km}$ $p_2V_2=nRT_2$. Hence, $nR$ remains constant over the ascent and 

$\displaystyle \frac{p_1V_1}{T_1}=\frac{p_2V_2}{T_2}$. Rearranging for the (required) volume at ground level $V_1$

$\displaystyle V_1=\frac{p_2V_2T_1}{p_1T_2}$

At $30\,\mathsf{km}$ the balloon is a sphere with diameter $20\,\mathsf{m}$ with volume 

$\displaystyle V_2=\frac{4}{3}\pi\left(\frac{20}{2}\right)^3$.

Using the data in the question, with temperatures as absolute temperatures in Kelvin

$\displaystyle V_1=\frac{4}{3}\pi\left(\frac{20}{2}\right)^3 \frac{1100(273+15)}{101.3\times10^3(273-40)}\,\mathsf{m}^3\frac{\mathsf{Pa\,K}}{\mathsf{Pa\,K}}$

$\implies \underline{V_1=56.2\,\mathsf{m}^3}$

Note that the volume at altitude $V_2\approx 4200\,\mathsf{m}^3$ - the gas has expanded around 75 times its original volume!


:::{hint} Concept: Thought experiment
:class: concept

**Straw with fluid**

What happens to the pressure in a straw when you place your finger over the top and try to pull the straw out of a glass of water?

**What forces are acting, and why doesn't the water simply fall out?**

::::{dropdown} See explanation

The water is held in the straw because the pressure above the water is lower than the atmospheric pressure acting at the bottom of the water column.

:::::{dropdown} Why is the pressure above the column lower than atmospheric pressure?

Take a look at the Perfect Gas Law. We have a fixed amount of air trapped between the water column and your finger. If the temperature is constant, the Perfect Gas Law tells us that

$
pV=nRT=\text{constant}.
$

Therefore, an increase in the volume of the trapped air must be accompanied by a decrease in its pressure.

When you first start to lift the straw, some of the water flows out of the bottom of the straw. This increases the volume of air trapped above the water column, causing its pressure to decrease. The volume increases until the pressure difference between the atmospheric pressure at the bottom of the column and the reduced pressure at the top is sufficient to support the weight of the water column.

The resulting force balance (for a straw with cross-sectional area $A$) is

$
(p_\mathrm{atm}-p_\mathrm{top})A=\rho ghA,
$

giving

$
p_\mathrm{atm}-p_\mathrm{top}=\rho gh.
$

So the pressure difference is exactly what is required to support the weight of the water column.


### Key assumptions

You can see from the Perfect Gas Law that pressure is not just linked to temperature, but also to density. The relationship between pressure and density depends on the **compressibility** of the fluid. For low to intermediate flow speeds, much less than the speed of sound in the fluid, we can usually assume that flows are **incompressible**, and the density is the same everywhere. We will explore compressible flows later in the course.

:::{hint} Concept: Compressibility
:class: concept

An incompressible fluid is one for which changes in pressure cause negligible changes in density.

:::

Furthermore, we typically assume that the rapid molecular-level fluctuations that give rise to the pressure and temperature in a fluid can be time-averaged by treating the fluid as a continuum.

:::{hint} Concept: Continuum assumption
:class: concept

The continuum assumption treats a fluid as a continuous medium, so that fluid properties such as pressure, temperature and density can be described by smoothly varying, time-averaged quantities rather than by the motion of individual molecules.

:::

### Surface tension

In a liquid the molecules are bound to each other by forces of molecular attraction (cohesion). At a boundary between two fluids this gives rise to "surface tension".

| Symbol | Quantity |  | SI unit |

Examples of surface tension: A fluid jet breaking up into droplets; A water-boatman using surface tension to support it's weight; Oil will not mix with water.  

Surface tension can also affect pressure. Although surface tension is introduced as an interfacial energy, it also represents a force acting along the interface. When the interface is curved, these surface-tension forces have a resultant normal to the interface. This must be balanced by a pressure difference across the interface.

For an interface with principal radii of curvature $R_1$ and $R_2$, the pressure jump is given by the Young-Laplace equation

$ \Delta p = \gamma \left( \frac{1}{R_1}+\frac{1}{R_2} \right). $

Notice that a flat interface has zero curvature, so surface tension does not produce a pressure difference across it. A curved interface, such as the surface of a droplet or bubble, does.

:::{prf:example} Pressure inside a water droplet
:enumerator: 5

A spherical water droplet has a radius of $1,\mathsf{mm}$. The surface tension of water is $\gamma=0.072\,\mathsf{N/m}$. If the pressure outside the droplet is atmospheric, $p_\mathrm{atm}=101.3\,\mathsf{kPa}$, determine the pressure inside the droplet.

::::{admonition} Solution
:class: dropdown

For a spherical interface, the Young-Laplace equation gives the difference in pressures inside and outside the droplet

$
\displaystyle \Delta p=\frac{2\gamma}{R} = \frac{2\times 0.072}{1\times10^{-3}}=144\,\mathsf{Pa}.
$

Therefore the pressure inside the droplet its this pressure difference above atmospheric pressure

$
p_{\mathrm{drop}}=p_\mathrm{atm}+\Delta p = 101.3+0.144\,\mathsf{kPa}
$

$
\implies \underline{p_{\mathrm{drop}}=101.444\,\mathsf{kPa}}.
$

The pressure inside the droplet is therefore slightly higher than the pressure outside because the surface tension of the curved interface must be balanced by a pressure difference.

::::

Because pressure fundamentally arises from molecular motion, zero pressure occurs when that motion ceases. That is, at absolute zero, zero Kelvin; a vacuum.

However, in engineering we usually deal with differences in pressure where the system is subject to the atmosphere. It is usually convenient to "zero" pressure measurements relative to atomspheric pressure. 

:::{tip} Definition: Absolute and gauge pressures
:class: definition

**Absolute pressure** is fluid pressure relative to zero pressure, a vacuum.

**Gauge pressure** is the fluid pressure relative to atmospheric pressure.

Hence, gauge pressure = absolute pressure - atmospheric pressure

:::

Gauge pressure is typically used in industry. If absolute pressure is $3\,\mathsf{bar}$ and atmospheric pressure is $1\,\mathsf{bar}$, then gauge pressure will be $2\,\mathsf{bar}$.

**Note that atmospheric pressure changes with altitude, temperature, weather!**

### Hydrostatic pressure

Fluid at rest has constant pressure across a **horizontal** plane - this is why liquid surfaces are flat and horizontal. But there **is** a pressure gradient in the vertical direction. So **how does pressure vary with depth in a fluid?** 

:::{hint} Concept: Thought experiment
:class: concept

If the bottle in the picture is filled with water so that a jet squirts out of each hole, which of the thress jets travels furthest? 

```{figure} images/jet_bottle1.png
:alt: Bottle with holes
:class: bg-none
:width: 20%
:align: left
```

::::{dropdown} See explanation

```{figure} images/jet_bottle.png
:alt: Bottle with jets squirting from holes
:class: bg-none
:width: 60%
:align: left
```

Water is pushed out of the hole by the weight of the fluid sitting above it. The bottom hole has the deepest column of fluid above it, so the jet here will have the highest velocity. The bottome jet will travel further than the jets from the holes higher up on the bottle. 

:::

It is the vertical gradient of **hydrostatic pressure** in the fluid that drives the jet out of the bottle.

Consider a small element of stationary fluid. The fluid is at rest, so there are no shear forces acting on the element, and the horizontal pressure is constant. However, the element has weight, $\delta m g$, acting vertically downwards. Because the element is stationary it must be in equilibrium $\rightarrow$ the sum of all the forces acting on the element must be zero, $\sum \text{Forces} = 0$.

The weight of the fluid element is counteracted by the difference in pressures acting vertically on the bottom and top surfaces of the element, as shown in the schematic. 

```{figure} images/fluid_element_def.png
:alt: Definition of a fluid element subject to a vertical pressure gradient
:class: bg-none
:width: 90%
:align: left
```

Considering the forces in the vertical, $z$, direction, we have
- weight $\delta m g$ acting downwards,
- a force due to the pressure acting **upwards** on the bottom surface of the element due to the water below it, $F_z = p \delta x \delta y$,
- a force due to the pressure acting **downwards** on the top surface of the element due to the water above it, $F_z +\delta F_z = (p + \delta p) \delta x \delta y$.

```{figure} images/element_balance.png
:alt: Vertical forces on a fluid element subject to a vertical pressure gradient
:class: bg-none
:width: 40%
:align: left
```

Resolving these vertical forces to ensure equilibrium, their sum must be zero with

$
F_z-(F_z+\delta F_z) - \delta mg = 0.
$

In terms of the element properties, from above

$
p \delta x \delta y - (p + \delta p) \delta x \delta y - g(\rho\delta x \delta y\delta z)=0.
$

This simplifies to

$
\delta p \delta x \delta y = - g(\rho\delta x \delta y\delta z),
$

and so 

$
\displaystyle \frac{\delta p}{\delta z} = - \rho g.
$

And so, in the limit as $\delta z \rightarrow 0$, the element becomes infinitessimally small and we form the differential eqution describing the vertical, **hydrostatic pressure** distribution in a stationary fluid

$
\displaystyle \frac{dp}{dz} = - \rho g.
$

For a homogeneous and incompressible fluid, density is assumed constant, and we can integrate to find the pressure difference between point 1 and point 2 in a fluid

$
\displaystyle \int_1^2\,dp = -\rho g\int_1^2\,dz \qquad \implies \qquad p_2-p_1 = -\rho g(z_2-z_1)
$

or

$
\Delta p = -\rho g\Delta z.
$

:::{hint} Take care of your coordinates
Note that here we have defined the vertical coordinate $z$ as positive upwards i.e. gravity and hence weight are negative quantities and hydrostatic pressure decreases with increasing $z$. Often you will see depth or elevation defined in the downwards direction, in the same direction as gravity. Clearly in that case, hydrostatic pressure increases with increasing depth $h$ following

$
\Delta p = \rho g\Delta h.
$

Essentially, it does not matter how you define these so long as you are consistent. **Always define your coordinates before working on a problem and stick to your definition**.

:::

(def-hydrostatic)=
:::{tip} Definition: Hydrostatic pressure
:class: definition

Hydrostatic pressure is the pressure within a fluid that is at rest, resulting from the weight of the fluid above the point of interest. Hence,

$
\displaystyle p=p_0 + \rho g h.
$

where $p_0$ is the pressure at the reference surface and $h$ is the vertical depth below that surface.



:::{prf:example} Hydrostatic pressure
:enumerator: 6

Atmospheric pressure of $1.01\,\mathsf{bar}$ acts on the surface of a lake $10\,\mathsf{m}$ deep. Find the absolute pressure at the bottom of the lake.

::::{admonition} Solution
:class: dropdown

If we denote the top of the lake to be at elevation $z_1$ and pressure $p_1=p_{\mathrm{atm}}$, and the bottom of the lake with elevation $z_2$ and pressure $p_2$, then the pressure at the bottom of the lake is due to the atmospheric pressure acting on the surface plus the hydrostatic pressure of the fluid above the base

$
p_2 = p_{\mathrm{atm}} + \rho_wg(z_1-z_2).
$

The difference in elevations $z_1-z_2=10\,\mathsf{m}$, hence

$
p_2 = 1.01\times 10^5 + 1000 \times 9.81 \times 10\,\mathsf{Pa}.
$

$
\implies \underline {p_2 = 199100\,\mathsf{Pa} =  199.1\,\mathsf{kPa} =  1.991\,\mathsf{bar}}.
$
:::{hint}
:class: dropdown

Remember to sense check your answer!
