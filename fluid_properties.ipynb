# Fluid properties

### Learning outcomes
- Define a fluid and its properties
- Understand the key engineering assumptions for describing a fluid
- Conceptualise the relationship between pressure and deformation
- Evaluate pressure distributions within a static fluid
- Recognise the difference between gauge and absolute pressures 
- Calculate fluid pressures from manometer readings

### What is a fluid?

- A fluid may be liquid, vapour or gas. It has no permanent shape but takes up the shape of a containing vessel or channel or is shaped by external forces (e.g.~the atmosphere).
- A fluid consists of atoms/molecules in random motion
and in continual collision with the surroundings.
- Fluids are readily deformable, and flow.

The key difference between liquids and gases is that where liquids have a fixed volume, gases may expand to fill container they are in.

### Shear in fluids

For a **solid**, application of a shear stress causes a deformation which, if modest (elastic), is not permanent. The solid regains original position when the stress is removed.

```{figure} images/shear_solid.png
:alt: Schematic showing the response of a solid to shear
:class: bg-none
:width: 80%
:align: center
```
In a **fluid** under shear stress, layers of molecules slide over each other leading to continuous deformation. This occurs until the stress is removed. 

```{figure} images/shear_fluid.png
:alt: Schematic showing the response of a fluid to shear
:class: bg-none
:width: 60%
:align: center
```

:::{hint} Concept: Fluid deformation
:class: concept

A fluid is a substance for which a shear stress results in unlimited, continuous deformation.
:::
