(section-submerged_surfaces)=
# Submerged surfaces

:::{seealso} 
### Recap of hydrostatic pressure

- In a continuous static fluid, pressure is constant in a horizontal plane
- The vertical pressure varies hydrostatically with

$
p=p_0+\rho g h
$

at a depth $h$ below a surface subject to a reference pressure $p_0$ in fluid of density $\rho$.

- We used these concepts to analyse how manometers measure pressure differences via differences in fluid column heights.

:::

:::{hint} Concept: Hydrostatic loads
:class: concept

- Hydrostatic pressure exerts forces on surfaces submerged in liquids at rest.
- Because the pressure is distributed over the submerged surface, the resulting force is also distributed over the surface.
- For design calculations we want the equivalent point force and its line of action.


Consider a plane surface submerged in a static fluid. We can calculate the resultant force by integrating the pressure over the surface. To locate its line of action, we must also calculate the moment of this distributed load.

:::

### What is the total force?

Imagine an arbitrary surface submerged in fluid with density $\rho$ and free surface pressure $p_0$. The surface has an element of area $\delta A$ at depth $h$ subject to hydrostatic pressure $p_0+\rho g h$. 

```{figure} images/hydrostatic_load_cartoon.png
:alt: Hydrostatic force on an arbitrary submerged surface
:class: bg-none
:width: 95%
:align: center
```
The force due to hydrostatic pressure acting on the element is

$ 
\delta F=(p_0+\rho gh)\,\delta A. 
$

To find the total force on the surface, we sum the contributions from every element - mathematically this means taking the limit as $\delta A \rightarrow 0$ to give a smooth integral

$
\displaystyle F=\int_A\,dF =\int_A(p_0+\rho gh)\,dA. 
$

If the same reference pressure $p_0$ acts on both sides of the surface, its contributions cancel. We can therefore calculate the net force using gauge pressure $p=\rho g h$, and 

$
\displaystyle F=\rho g\int_A h\,dA. 
$

This is the general expression for the resultant hydrostatic force on a surface.

What does that integral actually mean? 

:::{tip} Definition: Centroid
:class: definition

For a 3-D body the centre of gravity is the point at which all the mass acts. For a 2-D lamina (thin, flat plate) the **centroid** is the centre of area, the point about which a uniform lamina would balance.

If the submerged surface has area $A$ and $\tilde h$ is the vertical depth of its centroid below the fluid free surface, then

$
\displaystyle \tilde h=\frac{1}{A}\int_A h\,dA.
$

This is the area-weighted average depth of the surface.

:::

From this definition of the surface centroid, $\displaystyle \int_A h\,dA = A\tilde h$, giving us the hydrostatic load in terms of centroid position.


(highlight-hydrostatic_force)=
:::{important}

$
F=\rho g A\tilde h. 
$

The resultant hydrostatic force is therefore equal to the pressure evaluated at the centroid of the surface, multiplied by the surface area.

This is an exact result, not an approximation.

:::

::::{prf:example} Centroid of a rectangular plate
:enumerator: 11

Determine the location of the centroid of a plane rectangular plate.

```{figure} images/rect_centroid.png
:alt: Centroid of a rectangular plate
:class: bg-none
:width: 60%
:align: center
```

:::{admonition} Solution
:class: dropdown

As shown in the diagram, the centroid of the plate is at a location $(x_c,y_c)$, to be determined. To find the vertical coordinate $y_c$, we take moments about O-O of the horizontal element of area at $y$ and with thickness $\delta y$ and width $b$

$
\delta M = y\delta A = yb\delta y.
$

In the limit $\delta y\rightarrow 0$, we can integrate to sum the contributions to the total moment of all horizontal elements from the bottom of the plate ($y=0$) to the top ($y=d$), with 

$
\displaystyle M=b\int_0^d y\,dy = b\left[\frac{y^2}{2}\right]_0^d = \frac{bd^2}{2}.
$

But the total moment about O-O is the centroid vertical coordinate times the total area

$
M=bdy_c.
$

Equating, we find

$
\displaystyle y_c=\frac{1}{bd}\frac{bd^2}{2} = \frac{d}{2}.
$

Similarly, to find $x_c$, the second coordinate of the centroid position, we take moments about O'-O' of a vertical element of area at $x$ and with thickness $\delta x$ and width $d$

$
\delta M = x\delta A = xd\delta x.
$

In the limit $\delta x\rightarrow 0$, we can integrate to sum the contributions to the total moment of all horizontal elements from the left side of the plate ($x=0$) to the right ($x=b$), with 

$
\displaystyle M=d\int_0^b x\,dx = d\left[\frac{x^2}{2}\right]_0^b = \frac{db^2}{2}.
$

But the total moment about O'-O' is the centroid horizontal coordinate times the total area

$
M=bdx_c.
$

Equating, we find

$
\displaystyle x_c=\frac{1}{bd}\frac{db^2}{2} = \frac{b}{2}.
$

Therefore, the centroid of the rectangular plate is at 

$
\displaystyle \underline{(x_c,y_c)=\left(\frac{b}{2},\frac{d}{2}\right)}.
$

### Where does the resultant force act? 

The {ref}`expression above <highlight-hydrostatic_force>` gives the magnitude of the resultant force. We also need to know its line of action.

Considering the moment due to the **hydrostatic force** acting on the same element of surface we previously analysed

$
\delta M=h\,\delta F,
$

where $\delta F$ is hydrostatic force, $\delta F=\rho gh\,\delta A$.

In the limit $\delta A\rightarrow 0$, we form the smooth integral

$
\displaystyle M=\rho g\int_A h^2\,dA. 
$

However, if the resultant force acts at $h_{CP}$, then the moment is

$ 
\displaystyle M=Fh_{CP} = \rho gh_{CP}\int_A h\,dA. 
$

:::{tip} Definition: Centre of pressure
:class: definition

Equating these two expressions, we find

$
\displaystyle h_{CP} = \frac{\displaystyle\int_A h^2\,dA} {\displaystyle\int_A h\,dA}.
$

This is the **centre of pressure** of the surface. This is the point through which the resultant hydrostatic force acts.

The magnitude of the resultant force depends on the area of the surface and the depth of its centroid. The position of the centre of pressure also depends on the distribution of its area with depth.

### Horizontal surfaces

For horizontal surfaces submerged in a stationary fluid with no shear stresses, the hydrostatic forces - arising
from hydrostatic pressures - must act normal (i.e. perpendicular) to the surface.

With pressure constant in a horizontal plane, the pressure must be constant over the entire surface.

Therefore the **centre of pressure will coincide with centroid**.

The line of action of the resultant hydrostatic force is through the surface centroid. 

### Vertical surfaces

For a vertical plate pressure is not uniform, because pressure increases with depth. Therefore the centre of pressure does not coincide with the centroid. For a vertical plate, the centre of pressure is always below the centroid. In a stationary fluid, with no shear stresses, the resultant hydrostatic force acts normal to the plate through the centre of pressure.

::::{prf:example} Force on a vertical submerged surface
:enumerator: 12

The wall of a reservoir contains a vertical gate, $2\,\textsf{m}$ wide and $3\,\textsf{m}$ high. The upper edge of the gate is
level with the water surface. Taking $\rho = 1000\,\textsf{kg}/\textsf{m}^3$ and $g = 9.81\,\textsf{m}/\textsf{s}^2$, determine the net force
on the gate and the depth of the centre of pressure.

```{figure} images/wexp12.png
:alt: Vertical gate
:class: bg-none
:width: 60%
:align: center
```

::::{admonition} Solution
:class: dropdown


:::{hint} Strategy
 
1. Write expression for force on a small element of gate
2. Integrate to find total force
3. Write expression for moment for a small element about O-O, the reference depth
4. Integrate to find total moment
5. Equate moment for point force with moment for distributed force to find line of action

:::

1. The force on the small element of gate at depth $h$, with width $w$ and height $\delta h$

$
\delta F_p = \rho g h (w\delta h).
$

2. The total hydrostatic force is the integral between the bottom and top of the gate, $h_1=0\,\textsf{m}$ and $h_2=3\,\textsf{m}$

$
\displaystyle F_p=\rho g w \int_{h_1}^{h_2} h\,dh = \rho g w \left[\frac{h^2}{2}\right]_{h_1}^{h_2}=1000\times 9.81\times 2 \left[\frac{h^2}{2}\right]_{0}^{3}
$

$
\implies \underline{F_p = 9\rho g = 88.3\,\textsf{kN}}.
$

3. To find the line of action of this resultant force, find the moment of force on the small element

$
\delta M_{OO} = \delta F_p h = \rho g h (w\delta h)h.
$

4. The total moment is the integral between the bottom and top of the gate, $h_1=0\,\textsf{m}$ and $h_2=3\,\textsf{m}$

$
\displaystyle M_{OO}=\rho g w \int_{h_1}^{h_2} h^2\,dh = \rho g w \left[\frac{h^3}{3}\right]_{h_1}^{h_2}=1000\times 9.81\times 2 \left[\frac{h^2}{2}\right]_{0}^{3}
$

$
\implies M_{OO} = 18\rho g.
$

5. The moment due to distributed force must equal moment due to equivalent point force, therefore

$
F_ph_p = M_{OO}
$

$
\displaystyle \implies h_p=\frac{18\rho g}{9\rho g} = 2\,\textsf{m}.
$


::::{prf:example} Force on a vertical triangular gate
:enumerator: 13

A triangular aperture in a reservoir wall is sealed by a vertical gate. The gate is hinged along the bottom, horizontal edge and is kept closed by a horizontal force $F_N$ applied to the apex A. Find the required value of $F_N$ and the depth to the centre of pressure.

```{figure} images/wexp13.png
:alt: Vertical triangular gate
:class: bg-none
:width: 50%
:align: center
```

::::{admonition} Solution
:class: dropdown

:::{hint} Generic strategy

1. Note that width is a function of height $w= w(h)$ that we must identify
2. Write expression for force on a small element of gate
3. Integrate to find total force
4. Write expression for moment for a small element about O-O, the reference depth
5. Integrate to find total moment
6. Equate moment for point force with moment for distributed force to find line of action 
7. Take moments about the hinge to find $F_N$, the force to keep the gate shut

:::

1. Finding gate width as a linear funciton of its height, we can propose $w = k_1h + k_2$, with boundary conditions 

$
\text{when}\quad h = 4, w = 3 \quad\implies \quad 3 = 4k_1 + k_2
$

and
$
\text{when}\quad h = 1, w = 0 \quad\implies \quad 0 = k_1 + k_2.
$

Therefore $3=3k_1$ and so $k_1=1$ and $k_2=-1$ and

$
w=h-1
$

with units of metres. 

2. The force on the small element of gate at depth $h$, with width $w$ and height $\delta h$

$
\delta F_p = \rho g h (w\delta h) =  \rho g h (h-1)\delta h.
$

3. The total hydrostatic force is the integral between the bottom and top of the gate, $h_1=1\,\textsf{m}$ and $h_2=4\,\textsf{m}$

$
\displaystyle F_p=\rho g \int_{h_1}^{h_2} h(h-1)\,dh = \rho g \int_{h_1}^{h_2} (h^2-h)\,dh = \rho g \left[\frac{h^3}{3}-\frac{h^2}{2}\right]_{h_1}^{h_2}=1000\times 9.81 \left[\frac{h^3}{3}-\frac{h^2}{2}\right]_{1}^{4}
$

$
\implies \underline{F_p = 13.5\rho g = 132.4\,\textsf{kN}}.
$

4. Find the location of the centre of pressure $P$ by taking moments of the force on small element about O

$
\delta M_{O} = \delta F_p h = \rho g h (w\delta h)h = \rho g h (h-1) h\delta h= \rho g (h^3-h^2)\delta h.
$

5. The total moment is the integral between the bottom and top of the gate, $h_1=1\,\textsf{m}$ and $h_2=4\,\textsf{m}$

$
\displaystyle M_{O}=\rho g \int_{h_1}^{h_2} (h^3-h^2)\,dh = \rho g w \left[\frac{h^4}{4}-\frac{h^3}{3}\right]_{h_1}^{h_2}=1000\times 9.81\times 2 \left[\frac{h^4}{4}-\frac{h^3}{3}\right]_{1}^{4}
$

$
\displaystyle \implies M_{O} = 42\frac{3}{4}\rho g.
$

6. The moment due to distributed force must equal moment due to equivalent point force, therefore

$
F_ph_p = M_{O}
$

$
\displaystyle \implies \underline{h_p= 42\frac{3}{4}\frac{\rho g}{13.5\rho g} = 3.167\,\textsf{m}}.
$

7. The force required to keep the gate shut is $F_N$. Taking moments around the gate hinge

$
F_p(h_2-h_p) = F_N(h_2-h_1)
$

which can be rearranged for the closing force

$
\displaystyle F_N = \frac{F_p(h_2-h_p)}{(h_2-h_1)} = \frac{132.4(4-3.167)}{3}
$

$
\implies \underline{F_N=36.8\,\textsf{kN}}.
$

:::{note}
 
Was there a simpler way to deal with this geometry? 

You could directly use the centroid position of a triangle to calculate the resultant hydrostatic force. 

::::{note} 🔥 Generalisation to other surface geometries
:class: dropdown

**Inclined planar surfaces**

The same approach applies to a plane surface inclined at an angle to the horizontal. The pressure at each point still depends on its **vertical depth** below the free surface.

We can therefore calculate the resultant force from

$
F = \rho g A \bar h,
$

where $\bar h$ is the vertical depth of the surface centroid.

The centre of pressure can then be found by taking moments of the distributed load, just as we did for the vertical plate.

**Curved surfaces**

What about curved surfaces?

For a curved surface, the direction of the pressure force changes across the surface because pressure always acts normal to the surface. We therefore cannot generally find the resultant by simply adding the magnitudes of the elemental forces.

Instead, the pressure forces are resolved into horizontal and vertical components and each component calculated separately.

This is the approach used to determine hydrostatic loads on curved tank walls, dam gates and other submerged structures.

:::{note} Evaluating components

**Horizontal component**: equal to the hydrostatic force on the vertical projection of the curved surface.

**Vertical component**: equal to the weight of the imaginary volume of fluid directly above the curved surface, with the sign determined by the geometry.

This approach can also apply to inclined plane surfaces.

:::

**Second moment of area**

We don't necessarily want to integrate from scratch every time and sometimes geometries can become too complex to solve directly. Fortunately, the integrals appearing above are familiar quantities from engineering mechanics.

The first moment of area:
$
\displaystyle \int_A h\,dA=A\bar h 
$

and the second moment of area $I_G$ is defined
$
\displaystyle \int_A h^2\,dA = A\bar h^2+I_G. 
$

These are quantities that can be looked-up for many engineering-relevant shapes providing easy access to the resultant hydrostatic force

$
F=\rho gA\bar h 
$

and centre of pressure

$
\displaystyle h_{CP} = \bar h+\frac{I_G}{A\bar h}. 
$
