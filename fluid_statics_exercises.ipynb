### Buoyancy

:::{exercise}💧
The can shown in Figure Q1 floats in the position shown.  What is its weight in Newtons?  If the cylinder material is uniform what is the density of the cylinder?

```{figure} images/buoyant_cylinder.png
:alt: Floating cylinder
:class: bg-none
:width: 50%
:align: center
```
Figure Q1: A floating can, with dimensions, for use with exercise 1 (not to scale)


::::{dropdown} ✓ Check your answer
$4.99\,\textsf{N}$, $727\,\textsf{kg}/\textsf{m}^3$

:::{hint} Wrong density?
:class: dropdown

Have you calculated a density of $999\,\textsf{kg}/\textsf{m}^3$?
What is the height of the cylinder that gives it's weight?

::::

::::{admonition} Solution
:class: dropdown

The buoyancy force = weight of displaced water = immersed volume $\times \rho_w g$. Therefore, 

$
\displaystyle F_B = \frac{\pi}{4}d^2h_i\rho_w g = \frac{\pi}{4}0.009^2\times0.08\times1000\times9.81\,\textsf{m}^2\,\textsf{m}\,\textsf{kg}\,\textsf{m}^{-3}\,\textsf{m}\,\textsf{s}^{-2}
$

$
\implies \underline{F_B=4.99\,\textsf{N}}.
$

In equilibrium, the buoyancy force equals the weight of the cylinder, which is it's volume $V=h_c d^2 \pi/4$ multiplied by $\rho_c g$. Hence

$
F_B= \rho_c g h_c d^2 \pi/4.
$

Rearranging for the cylinder density

$
\displaystyle \rho_c = \frac{4F_B}{\pi g h_c d^2} =  \frac{4\times 4.99}{\pi \times 9.81\times 0.11 \times 0.09^2}\,\frac{\textsf{N}}{\textsf{m}\,\textsf{s}^{-2}\,\textsf{m}\,\textsf{m}^2}
$

$
\implies \underline{\rho_c=727\,\textsf{kg}/\textsf{m}^3}.
$


::::

:::{exercise}💧
A spherical plastic fishing float of mass $28\,$g is to be used for seawater fishing.  The bait attached to the float has a mass of $7\,$g and a volume of $4\times10^{-6}\,\textsf{m}$$^3$.  If half of the float is visible above the water surface what is the diameter of the float? 

The density of sea water is $1025\,\textsf{kg}/\textsf{m}^3$.

::::{dropdown} ✓ Check your answer
$48.7\,\textsf{m}$m
::::

::::{admonition} Solution
:class: dropdown


```{figure} images/float.png
:alt: Float and bait
:class: bg-none
:width: 50%
:align: center
```
The float and bait are floating in equilibrium, so the buoyancy force $F_B$ balances the weight $W=mg$. The buoyancy force is $\rho g$ times the displaced volume

$
\displaystyle F_B=\rho g\left(V_{\text{bait}}+\frac{1}{2}V_{\text{sphere}}\right)
$

or

$
\displaystyle F_B=\rho g\left(V_{\text{bait}}+\frac{1}{2}\frac{\pi}{6}d^3\right).
$

Equating $F_B=W$

$
\displaystyle mg = \rho g\left(V_{\text{bait}}+\frac{\pi}{12}d^3\right),
$

which, cancelling through by $g$, can be rearranged for the sphere diameter $d$

$
\displaystyle d= \left[\frac{12}{\pi}\left(\frac{m}{\rho} - V_{\text{bait}}\right)\right]^{1/3} = \left[\frac{12}{\pi}\left(\frac{(0.028+0.007)}{1025} - 4\times10^{-6}\right)\right]^{1/3},
$

$
\implies \underline{d= 48.7\,\textsf{mm}}.
$

:::{exercise}📝
A barge, $l=15\,\textsf{m}$ long and $h=3\,\textsf{m}$ deep, has a trapezoidal cross-section which is $b=9\,\textsf{m}$ wide at the top and $a=6\,\textsf{m}$ wide at the bottom.  Both ends of the barge are vertical.  The weight of the barge empty is $2.0\,\textsf{m}$N and it carries an evenly distributed load of coal of mass $75\,$tonnes.  The barge floats horizontally at rest in sea water of density $1025\,\textsf{kg}/\textsf{m}^3$.  To what depth is the barge immersed in the water?

::::{dropdown} ✓ Check your answer
$2.502\,\textsf{m}$
::::

::::{admonition} Solution
:class: dropdown

```{figure} images/barge_soln.png
:alt: Trapezoidal barge
:class: bg-none
:width: 60%
:align: center
```
Let the barge be immersed to a depth $d$, measured from the bottom of the barge. Recalling that the buoyancy force is equal to the weight of displaced water

$
\displaystyle F_B=\frac{1}{2}(a+w)dl\rho_w g
$

with the width of the barge at the water-surface level $w$ a function of $d$, $w=k_1d+k_2$.

Applying boundary conditions

$
\text{when}\quad w=6, d=0\quad \implies \quad 6=0+k_2\quad \implies \quad k_2=6
$

and 

$
\text{when}\quad w=9, d=3\quad \implies \quad 9=3k_1+k_2\quad \implies \quad k_1=1,
$

leading to $w=d+k_2$ (with units of metres) and 

$
\displaystyle F_B=\frac{1}{2}(a+d+k_2)dl\rho_w g.
$

Applying equilibrium, this buoyancy force equates to the weight of the barge plus its load, with

$
\displaystyle  \frac{1}{2}(a+d+k_2)dl\rho_w g =  \frac{1}{2}(12+d)15d\times1025 g = 2\times10^6 + 75\times10^3g.
$

This gives us a quadratic equation in $d$, with coefficients we can evaluate

$
d^2 + 12d -36.28 = 0
$

which can be solved by appying the quadratic formula

$
\implies\underline{d=2.502\,\textsf{m}}.
$

:::{hint} Quadratic solutions
:class: dropdown

Quadratic equations can have two solutions, which this one does. However, the other solution here is $-14.5\,\textsf{m}$ i.e. it is negative, which would mean the barge would be levitating above the water! Clearly this is physically not possible, so we know that the crrent solution has to be the positive valued one. 

:::

::::

:::{exercise}📝
A buoy of mass $25\,$kg and diameter $1.2\,\textsf{m}$ is connected to the bottom of a seawater harbour. A $2.3\,$kg bird sits on top of the buoy.  What is the tension in the cable if half of the buoy is below water?  Take the density of sea water to be $1025\,\textsf{kg}/\textsf{m}^3$.


::::{dropdown} ✓ Check your answer
$4281\,\textsf{N}$
::::

::::{admonition} Solution
:class: dropdown

```{figure} images/buoy_soln.png
:alt: Thethered buoy
:class: bg-none
:width: 50%
:align: center
```

The buoy is in equilibrium therefore the buoyancy force counteracts the total weight

$
F_B=W_{\text{TOT}}+T.
$

The buoyancy force is also the weight of displaced water

$
F_B=\rho V_{\text{sub}} g.
$

Equating these expressions

$
\rho V_{\text{sub}} g = W_{\text{TOT}}+T = m_{\text{TOT}}g+T,
$

which can be rearranged for the tension

$
\displaystyle T = g(\rho V_{\text{sub}} - m_{\text{TOT}}) = 9.81\left[\frac{1025}{2}\frac{\pi 1.2^3}{6} - (25+2.3)\right]
$

$
\implies \underline{T=4281\,\textsf{N}}.
$

What is the tension, $T$, in the string?

Figure Q5: A cube suspended within a two-layer oil-water stratification (not to scale)

::::{dropdown} ✓ Check your answer
$48.1\,\textsf{N}$
::::

::::{admonition} Solution
:class: dropdown

Here the buoyancy force plus the cable tension counteract the weight of the cube

$
F_B+T=W=mg
$

where the buoyancy force is the weight of displaced fluid

$
\displaystyle F_B=\frac{1}{2}l^3\left(\rho_w+\rho_o\right)g.
$

Equating $F_B$

$
\displaystyle \frac{1}{2}l^3\left(\rho_w+\rho_o\right)g = mg - T,
$

which is rearranged to give the tension

$
\displaystyle T = g\left[m - \frac{1}{2}l^3\left(\rho_w+\rho_o\right)\right] = 9.81\left[12 - \frac{1}{2}0.2^3\left(1025+750\right)\right],
$

$
\implies\underline{T=48.1\,\textsf{N}.}
$
::::

### Hydrostatic forces on submerged surfaces 

:::{exercise} 💧
A thin rectangular plate sits suspended horizontally in fresh water at a depth of $10\,\textsf{m}$.  The plate has a width of $3\,\textsf{m}$ and length $5\,\textsf{m}$.
1. What is the gauge pressure acting on the surface? 
2. What is the hydrostatic force due to the gauge pressure on the upper face of the plate?  
3. What is the position of the point of action of that hydrostatic force?

::::{dropdown} ✓ Check your answer
1. $98.1\,\textsf{kPa}$
2.  $1.47\,\textsf{MN}$
3. [widthwise, lengthwise], [$1.5\,\textsf{m}$, $2.5\,\textsf{m}$]
::::

::::{admonition} Solution
:class: dropdown
1. Gauge pressure is the pressure relative to atmospheric, so in this case the pressure caused by the depth of fluid above the plate, so

$
p=\rho g h = 1000\times 9.81 \times 10\,\textsf{Pa}
$

$
\implies \underline{p=98.1\,\textsf{kPa}}
$

2. By definition, "Pressure = Force/Area", hence we can evaluate the force acting on the horizontal surface

$
F=pA = \rho g h A = (1000\times 9.81 \times 10)(3\times 5)\,\textsf{N}
$

$
\implies \underline{F=1.47\,\textsf{MN}}
$


3. The force acts at the centre of area, which for a rectangular system is a simple analysis (see notes). So a point at half the width and half the length; widthwise at $3/2 = 1.5\,\textsf{m}$ and lengthwise at $5/2 = 2.5\,\textsf{m}$.

::::

:::{exercise} 📝
A thin plate in the shape of a trapezium of top edge $15\,\textsf{m}$, bottom edge $20\,\textsf{m}$ and height $10\,\textsf{m}$ is suspended horizontally in sea water of density $1025\,\textsf{kg}/\textsf{m}^3$ at a depth of $4\,\textsf{m}$ (i.e. the entire face of the trapezium is $4\,\textsf{m}$ below the surface of the water and parallel to it).

What is the hydrostatic force due to the gauge pressure on the upper face of plate and its line of action.

::::{hint} Hint
:class: dropdown

The area of a trapezium is 

$\displaystyle \frac{1}{2}(a+b)c$.

The centroid (centre of area) of a triangle is $1/3$ of the way along a line joining the vertex to the base.  You can find the centroid of the trapezium by thinking of it as a rectangle and two triangles.

```{figure} images/trapezium_info.png
:alt: Trapezium geometry
:class: bg-none
:width: 90%
:align: center
```


The line of action is at the centroid of the plate. Take moments for area about OO. Split trapezium into two triangles (centroid 1/3 height from base) and a rectangle (centroid at centre) giving

$
\displaystyle 2(\textsf{Area of triangle})\frac{h}{3}+(\textsf{Area of rectangle})\frac{h}{2} = (\textsf{Area of trapezium})x_p.
$

$
\displaystyle 2\left(\frac{1}{2}\times 2.5\times 10\right)\frac{10}{3}+(15\times 10)\frac{10}{2} = \left[\frac{1}{2}(15+20)10\right]x_p.
$

$
\implies \underline{x_p=4.76\,\textsf{m}}.
$

::::

:::{exercise} 📝
An open topped rectangular tank is completely filled with oil of Relative Density (density relative to water, and also known as Specific Gravity) $0.87$. The tank is $6\,\textsf{m}$ long, $4\,\textsf{m}$ wide and $4\,\textsf{m}$ high.  

Calculate:
1. the net force on the base of the tank;
2. the net force on one of the square end panels;
3. the line of action of the force on the square end panels.


::::{dropdown} ✓ Check your answer
1. $819\,\textsf{kN}$
2. $273\,\textsf{kN}$
3. $2.67\,\textsf{m}$ below the surface
::::

::::{admonition} Solution
:class: dropdown
1. At the base of the tank 

$
F=\rho g h A = (0.87\times 1000\times 9.81\times 4)(4\times 6)
$

$
\implies \underline{F=819\,\textsf{kN}}.
$

2. For a small horizontal element in an end panel

$
\delta F = \rho g h \delta A = \rho g h w \delta h.
$

Taking the limit as $\delta h \rightarrow 0$, we can integrate

$
\displaystyle F=\rho g w \int_0^h h\,dh = 870\times 9.81 \times 4\left[\frac{h^2}{2}\right]_0^4 = 32\rho g
$

$
\implies \underline{F=273\,\textsf{kN}}.
$

:::{hint} Hint
:class: dropdown

For this linear hydrostatic pressure distribution, you can also consider the force on the end wall to be the **average** force exerted due to the hydrostatic pressure, which is the pressure at the height of the wall centroid $h_c=2\,\textsf{m}$. 

$
F'=\rho g h_c A = 870 \times 9.81 \times 2(4\times 4) = \underline{32\rho g = 273\,\textsf{kN}}.
$

:::

3. Taking moments about the surface, for a small element

$
\delta M = \delta F h = 4\rho g h^2\,\delta h.
$

Taking the limit as $\delta h \rightarrow 0$, we can integrate

$
\displaystyle M=4\rho g\int_0^h h^2\,dh = 4\rho g\left[\frac{h^3}{3}\right]_0^4 = 85.3\rho g.
$

Equating point and distributed moments

$
M=Fy_p,
$

hence $\displaystyle y_p=\frac{85.3\rho g}{32 \rho g}$

$
\implies \underline{y_p=2.67\,\textsf{m}}.
$

That is, the net force acts at a distance $2.67\,\textsf{m}$ below the oil surface. 

::::

:::{exercise} 📝
A vertical, rectangular gate in the side of a lock is $1\,\textsf{m}$ high and $2\,\textsf{m}$ wide.  The gate is hinged along its top edge and the depth of water to the base of the gate is $3\,\textsf{m}$.  

Calculate:	
1. the net hydrostatic force exerted on the gate (due to pressure);
2. the line of action of that force;
3. the moment of that force about the hinge.

::::{dropdown} ✓ Check your answer
1. $49.05\,\textsf{kN}$ 
2. $2.53\,\textsf{m}$ below water surface
3. $26.16\,\textsf{kN}\,\textsf{m}$
::::

::::{admonition} Solution
:class: dropdown
1. At depth $h$ the guage pressure is $\rho g h$. The hydrostatic force on a section of gate with width $w$ and height $\delta h$ is

$
\delta F = \rho g h \delta A = \rho g h w \delta h.
$

The total force is found by taking the limit as $\delta h \rightarrow 0$ and integrating over the gate height

$
\displaystyle F = \rho g w \int_{h_1}^{h_2} h\,dh = \rho g w \left[ \frac{h^2}{2}\right]_{h_1}^{h_2}
$

$
\displaystyle F = 1000\times 9.81\times 2\left[\frac{h^2}{2}\right]_2^3 1000\times 9.81\times 2\left[\frac{9}{2}-\frac{4}{2}\right]
$

$
\implies\underline{F = 49.05\,\textsf{kN}}
$

2. To find the line of action, take moments about the water surface. For an element at depth $h$

$
\delta M = h\delta F = (\rho g h)(w \delta h)h.
$

Taking limits and integrating

$
\displaystyle M = \rho g w \int_{h_1}^{h_2} h^2\,dh = \rho g w \left[\frac{h^3}{3}\right]_{h_1}^{h_2}.
$

Evaluating

$
\displaystyle M = 1000\times 9.81\times 2\left[\frac{h^3}{3}\right]_2^3 1000\times 9.81\times 2\left[\frac{27}{3}-\frac{8}{3}\right]= 124.26\,\textsf{kN}\,\textsf{m}.
$

But $M=Fh_p$, thus the line of action acts at 

$
\displaystyle\implies \underline {h_p = \frac{M}{F} = \frac{124.26}{49.05} = 2.533\,\textsf{m},}
$

that is, the force acts $2.533\,\textsf{m}$ below the water surface. 

```{figure} images/moments_soln.png
:alt: Taking moments about the hinge
:class: bg-none
:width: 60%
:align: center
```

3. Taking moments about the hinge

$
M = F (h_p-h_1) = 49.05\times 0.53
$

$
\implies \underline{M = 26.0\,\textsf{kN}\,\textsf{m}}
$

Alternatively, the moment about the hinge can be found directly, considering an element at depth $h$, where the moment of hydrostatic force about the hinge is 

$
\delta M = h\delta F = (\rho g h)(w \delta h)h_h,
$

where $h_h$ is the distance from the hinge, that is $h_h = h-h_1$. So

$
\delta M = h\delta F = (\rho g h)(w \delta h)(h-h_1),
$

when taking limits and integrating

$
\displaystyle M = \rho g w \int_{h_1}^{h_2} h(h-h_1)\,dh = \rho g w \left[\frac{h^3}{3}-\frac{h_1h^2}{2}\right]_{h_1}^{h_2}.
$

Evaluating

$
\displaystyle M = 1000\times 9.81\times 2\left[\frac{h^3}{3}-h^2\right]_2^3 1000\times 9.81\times 2\left[(9-9)-(\frac{8}{3}-4)\right]= 26.16\,\textsf{kN}\,\textsf{m}.
$

:::{exercise} 📝
A vertical gate in a reservoir wall has the form of an isosceles triangle, as indicated in Figure Q11.  The upper edge of the gate is horizontal, $3\,\textsf{m}$ wide and lies $3\,\textsf{m}$ below the water surface.  The gate is hinged about a horizontal axis at a distance $x$ from the top
of the gate.  The gate is prevented from rotating in one direction by a stop.  It is required that the gate is on the point of opening
automatically when the water is $5\,\textsf{m}$ deep.  

Neglecting friction at the hinge, find $x$.  

```{figure} images/vertical_gate.png
:alt: Vertical reservoir gate
:class: bg-none
:width: 80%
:align: center
```
Figure Q11: A vertical gate in a reservoir, with dimensions, for use with exercise11 (not to scale)


::::{hint} Hint
:class: dropdown
Think about where the hinge would need to be with respect to the resultant force for the gate to open.
::::

::::{dropdown} ✓ Check your answer
$8/11\,\textsf{m}$ $= 0.727\,\textsf{m}$
::::

::::{admonition} Solution
:class: dropdown

```{figure} images/moments2_soln.png
:alt: Taking moments about the hinge
:class: bg-none
:width: 60%
:align: center
```

There are two ways of solving this problem! The first is a bit easier. 

**Centre of pressure approach**

If the line of action of the hydrostatic pressure force is at the hinge, then there is no resultant moment about the hinge and the gate must be on the point of opening. 

The net force acting on the gate at the centre of pressure is found from considering an element at depth $h$ with

$
\delta F =\rho g h w \delta h.
$

We need to find the width $w$ as a function of $h$ which, if we write $w=k_1h + k_2$, the boundary conditions give us

$
\textsf{when}\quad h=3, w=3 \quad \implies 3=3k_1+k_2 
$

and 

$
\textsf{when}\quad h=5, w=0 \quad \implies 0=5k_1+k_2.
$

Subtracting gives us

$
\displaystyle k_1 = \frac{-3}{2},\quad k_2=-5k_1 = \frac{15}{2} \quad \implies \quad w = \frac{3}{2}\left(5-h\right).
$

This gives us the hydrostatic force on an element of fluid at depth $h$

$
\displaystyle \delta F =\rho g h w \delta h = \frac{3}{2}\rho g h(5-h) \delta h.
$

Taking the limit and integrating

$
\displaystyle F =\frac{3}{2} \rho g\int_{h_1}^{h_2} h(5-h)\,dh = \frac{3}{2}\rho g\left[\frac{5h^2}{2}-\frac{h^3}{3}\right]_3^5 = 11\rho g
$

We can find the moment due to the hydrostatic force by taking moments about 00 (see schematic), for an element of surface at depth $h$ from the hinge

$
\displaystyle \delta M = Fh = \left[\frac{3}{2}\rho g h(5-h) \delta h\right]h.
$

Integrating

$
\displaystyle M_{00} =  \frac{3}{2}\rho g \int_{h_1}^{h_2} h^2(5-h) \delta h = \frac{3}{2}\rho g \left[\frac{5h^3}{3}-\frac{h^4}{4}\right]_3^5 =41\rho g.
$

Given $M_{00}=Fy_p$

$
\displaystyle y_p=\frac{M_{00}}{F}=\frac{41\rho g}{11\rho g}==3.727\,\textsf{m}
$

The net hydrostatic force acts at a distance of $3.727\,\textsf{m}$ below the water surface. If the hinge is located exactly at this depth then there is no resultant hydrostatic force on the gate and it will be on the point of opening. In this case

$
\displaystyle \implies \underline{x=y_p-3 = \frac{41-33}{11} =  \frac{8}{11} = 0.727\,\textsf{m}.}
$

**Direct approach, taking moments**

Taking moments about the hinge - let the hinge be located at a depth $y_h$ below the water surface. As before, at a depth $h$ the pressure is $\rho g h$ and the force acting on an element of height $\delta h$ is

$
\displaystyle \delta F =\rho g h w \delta h = \frac{3}{2}\rho g h(5-h) \delta h.
$

The distance of the element from the hinge is $h-y_h$, so the moment of the force acting on the element is

$
\displaystyle \delta M = \frac{3}{2}\rho g h(5-h)(h-y_h) \delta h.
$

When the gate is on the point of opening, the total moment will be zero. So taking the limit and integrating over the gate will equate to zero

$
\displaystyle M = \frac{3}{2}\rho g \int_{h_1}^{h_2} h(5-h)(h-y_h)\,dh = \frac{3}{2}\rho g \int_{h_1}^{h_2} (5h^2 - h^3 - 5hy_h + h^2y_h)\,dh
$

leading to
$
\displaystyle M = \frac{3}{2}\rho g \left[\frac{5h^3}{3} - \frac{h^4}{4} - \frac{5h^2y_h}{2} + \frac{h^3y_h}{3}\right]_{h_1}^{h_2}=0.
$

Hence, the contents of the big bracket, evaluated at $h_1=3\,\textsf{m}$ and $h_2=5\,\textsf{m}$ is zero

$
\displaystyle \left[\frac{5h^3}{3} - \frac{h^4}{4} - \frac{5h^2y_h}{2} + \frac{h^3y_h}{3}\right]_{h_1}^{h_2}=0.
$

Therefore

$
\displaystyle y_h=\frac{41}{11} = 3.727\,\textsf{m},
$

and

$
\implies \underline{x=y_h-3 = 0.727\,\textsf{m}.}
$

A tank with vertical sides contains water to a depth of $1.5\,\textsf{m}$ and a layer of oil $0.6\,\textsf{m}$ resting on top of the water.  In one vertical side of the tank a rectangular hole of width $0.8\,\textsf{m}$ and extending the full height of the tank has been covered exactly by a flat plate to prevent leakage.  The relative density of the oil is $0.9$ and the air above the oil and surrounding the outside of the tank is at atmospheric pressure.  

Calculate the net hydrostatic thrust on the plate and the distance of the line of action of the thrust above the base of the tank.

::::{hint} Hint
:class: dropdown
When considering the water, remember the pressure at the top is not atmospheric.  Therefore the pressure in the water is not simply $\rho g h$.
::::

::::{dropdown} ✓ Check your answer
$16.457\,$kN, $0.69\,\textsf{m}$
::::

::::{admonition} Solution
:class: dropdown
Considering first the oil layer, the hydrostic force applied to an element of the plate with width $w=0.8\,\textsf{m}$ and height $\delta h$ at depth $h$ is

$
\delta F_o = \rho_o g h w \delta h.
$

Taking the limit and integrating

$
\displaystyle F_o = w\rho_o g\int_{h_1}^{h_2} h\,dh = w\rho_o g\left[\frac{h^2}{2}\right]{h_1}^{h_2},
$

and evaluating

$
\displaystyle F_o = 0.8\rho_o g\left[\frac{0.6^2}{2}\right] = 0.18(0.8\rho_o g)\,\textsf{N}.
$

The moment of that element of surface

$
\delta M_o = \rho_o g h w \delta h h
$

which integrates

$
\displaystyle M_o = w\rho_o g\int_{h_1}^{h_2} h^2\,dh = w\rho_o g\left[\frac{h^3}{3}\right]{h_1}^{h_2}=0.072(0.8\rho_o g)\,\textsf{N}.
$

$
\displaystyle M_o=F_oh_p\quad \implies \quad h_p=\frac{0.072}{0.18} = 0.4\,\textsf{m}.
$

Repeat this process for the water, remembering that the oil is producing a pressure on the surface of the water. We consider an element at $y$ below the water surface, where $h_o$ is the depth of the oil layer.

$
\delta F_w = (h_o\rho_o g + \rho_w g y)w\delta y,
$

which gives 

$
\displaystyle F_w = w g\int_{y_1}^{y_2} h_o\rho_o+\rho_w y\,dy = w g\left[h_o\rho_o y+\rho_w\frac{y^2}{2}\right]{y_1}^{y_2}.
$

Evaluating,

$
\displaystyle F_w = 0.8\times 9.81\left[0.6\times900 y+1000\frac{y^2}{2}\right]{0}^{1.5} = 15186\,\textsf{N}.
$

Now finding the moment of the element

$
\delta M_w = (h_o\rho_o g h +\rho_w g y) w \delta y y.
$

Taking the limit

$
\displaystyle M_w = w g\int_{y_1}^{y_2} (h_o\rho_o y+\rho_w y^2)\,dy,
$

which can be integrated and evaluated

$
\displaystyle M_w = w g\left[h_o\rho_o \frac{y^2}{2}+\rho_w\frac{y^3}{3}\right]{y_1}^{y_2} = 0.8\times 9.81\left[0.6\times900\frac{y^2}{2}+1000\frac{y^3}{3}\right]{0}^{1.5},
$

$
\implies M_w = 13596.7\,\textsf{kN}\,\textsf{m}.
$

But 

$
\displaystyle M_w = F_wy_p \quad \implies \quad y_p=\frac{13596.7}{15186} = 0.895\,\textsf{m}.
$

The total, resultant force is the sum of the oil layer and water layer contributions

$
\implies \underline{\mathcal{F} = F_o+F_w = 1271.4 + 15186 = 16.457\,\textsf{kN}.}
$

Taking moments about the top of the oil

$
\mathcal{F}x = F_oh_p + F_w(y_p+h_o) = (1271.4\times0.4)+(15186\times1.495)
$

$
\implies\underline{x=1.41\,\textsf{m}}
$

from the top of the tank.  

Therefore the resultant force acts at $0.69\,\textsf{m}$ from the base of the tank.

**Indirect approach (using trigonometry)**

Applying trigonometry to the geometry shown in the figure, we can find the vertical and horizontal projections of the gate dimensions

$
H=0.8\sin38=0.49\,\textsf{m} \quad\text{and}\quad W=0.8\cos38=0.63\,\textsf{m}.
$

The force on an equivalent vertial gate, width $w$

$
\displaystyle F_H=\rho g w\int_{h_1}^{h_2} h\,dh = \rho g w\left[\frac{h^2}{2}\right]_{h_1}^{h_2}
$

where $h_1$ and $h_2$ are $1.7\,\textsf{m}$ and $2.19\,\textsf{m}$ respectively.

Evaluating

$
F_H=0.57\rho g = 4488\,\textsf{N}.
$

This is the horizontal component of the perpendicular force applied with

$
F_H=F_P\sin38\quad \implies \quad F_P = 7290\,\textsf{N}.
$


Alternatively, we can find the total force by analysing the vertical force $F_v$, which is the weight of water above the gate

$
F_V = w\left[(h_1W) + (0.5WH)\right]\rho g = 0.6\left[(1.7\times0.63) + (0.5\times0.63\times0.49)\right]800\times9.81, 
$

$
F_V=0.74\rho g = 5770\,\textsf{N}.
$

The resultant force therefore has magnitude

$
\displaystyle \underline{F_P=\sqrt{F_H^2+F_V^2} = 7310\,\textsf{N}.}
$

The force acts perpendicular to the gate, with the point of action found by considering the equivalent vertical gate and taking moments about the surface. In integral form

$
\displaystyle M=\rho g w\int_{h_1}^{h_2} h^2\,dh = 0.6\rho g\left[\frac{h^3}{3}\right]_1.7^2.19 = 1.12\rho g.
$

Hence, with $F_Hy_p=M$

$
\displaystyle \implies \underline{y_p=\frac{1.12\rho g}{0.57\rho g}=1.96\,\textsf{m}.}
$

A force of $7310\,\textsf{N}$ acts perpendicular to the gate at $1.96\,\textsf{m}$ below the surface. 

**Direct approach (tanking moments)**

The force on an element of gate

$
\delta F = \rho g h w \delta y 
$

where $y$ is the spatial coordinate parallel to the gate surface (see figure). Noting that $h=h(y)$ we can write this function by applying boundary conditions to the linear relationship $h=k_1y+k_2$: When $h=1.7, y=0$ and when $h=2.19, y=0.8$. Hence, 

$
1.7 = 0+k_2\quad\implies\quad k_2=1.7,
$

$
2.9 = 0.8k_1+1.7\quad\implies\quad k_2=0.613
$

and $h=0.613y+1.7$.

Taking limits and integrating for the force

$
\displaystyle F=\rho g w\int_{y_1}^{y_2} k_1y+k_2\,dy = \rho g w\left[k_1\frac{y^2}{2}+k_2y\right]_{y_1}^{y_2}. 
$

Using the values evaluated for $k_1$ and $k_2$, and the data provided, we can evaluate

$
\displaystyle F= \rho g 0.6\left[0.613\frac{y^2}{2}+1.7y\right]_{0}^{0.8} = 0.934\rho g
$

$
\implies\underline{F=7330\,\textsf{N}}.
$

Taking moments about the end of the gate, acting on an element of gate

$
\delta M = \delta F y = \rho g w (k_1y+k_2)y\,dy.
$

Taking limits and integrating for the moment

$
\displaystyle M=\rho g w\int_{y_1}^{y_2} k_1y^2+k_2y\,dy = \rho g w\left[k_1\frac{y^3}{3}+k_2\frac{y^2}{2}\right]_{y_1}^{y_2}. 
$

Using the values evaluated for $k_1$ and $k_2$, and the data provided, we can evaluate

$
\displaystyle M= \rho g 0.6\left[0.613\frac{y^3}{3}+1.7\frac{y^2}{2}\right]_{0}^{0.8} = 0.389\rho g.
$

Equating the moment with the force through the centre of pressure, $Fy_p=M$ gives us 

$
\displaystyle y_p=\frac{M}{F}=\frac{0.389\rho g}{0.934\rho g}=0.416\,\textsf{m}.
$

Relative to the fluid surface $h_p=k_1y_p+k_2 = 0.613y_p+1.7 1.94\,\textsf{m}$.

Hence, a force of $7330\,\textsf{N}$ acts $0.416\,textsf{m}$ from the upper end of the gate, perpendicular to the gate.

The submerged trapezoidal face of a dam wall slopes at $60^\circ$ to the horizontal.  It is $180\,\textsf{m}$ long at its base and $300\,\textsf{m}$ long at the water line when the water is $30\,\textsf{m}$ deep. Find
1. The net force on the wall and 
2. The depth of the centre of pressure.

::::{dropdown} ✓ Check your answer
1. $1.12\,\textsf{GN}$
2. $19.1\,\textsf{m}$
::::

::::{admonition} Solution
:class: dropdown

```{figure} images/trap_inc_soln.png
:alt: Trapezoidal dam submerged
:class: bg-none
:width: 80%
:align: center
```

**Indirect approach**

Consider the equivalent vertical gate, at trapezium of top $300\,\textsf{m}$, bottom $180\,\textsf{m}$ and height $30\,\textsf{m}$ and calculate horizontal component of force on an element

$
\delta F = \rho g h w \delta h.
$

Finding $w$ as a function of $h$, we pose that $w=k_1h+k_2$ and apply boundary conditions

$
w=300\quad \text{when}\quad h=0 \quad \implies \quad k_2=300,
$

$
w=180\quad \text{when}\quad h=30 \quad \implies \quad k_1=-4,
$

so $w=300-4h$ (with units of metres).

Hence, taking limits and integrating, the horizontal component of force 

$
\displaystyle F_H=\rho g\int_{h_1}^{h_2} h(k_1h+k_2)\,dh =\rho g\left[\frac{k_1h^3}{3}+k_2\frac{h^2}{2}\right]_{h_1}^{h_2}
$

where $h_1=0$ and $h_2=30\,\textsf{m}$. Evaluating

$
F_H=99000\rho g =971\,\textsf{MN}.
$

Calculating $F_V$ via the fluid weight is very challenging for this geometry! However, the net force acts perpendicular to the gate and the centre of pressure is on the gate. Line of action can therefore be found by considering the equivalent vertical gate. Taking moments of an element about the water surface

$
\delta M=\rho g h^2 w \delta h. 
$

Taking limits and integrating, the moment

$
\displaystyle M=\rho g\int_{h_1}^{h_2} h^2(k_1h+k_2)\,dh =\rho g\left[\frac{k_1h^4}{4}+k_2\frac{h^3}{3}\right]_{h_1}^{h_2}
$

where $h_1=0$ and $h_2=30\,\textsf{m}$. Evaluating

$
M=1.89\times10^6\rho g\,\textsf{N}.
$

Given $M=Fh_p$

$
\displaystyle \implies\underline{h_p = \frac{1.89\times10^6}{9.9\times10^4} = 19.1\,\textsf{m}.}
$

**Direct approach**

At depth $h$ the hydrostatic pressure is $\rho g h$ and the corresponding force on a small element of the dam wall is

$
\delta F = \rho g h w \delta y
$

where $w$ and $h$ both vary with $y$, with

$
\displaystyle h=y\sin60 = \frac{\sqrt{3}}{2}y,
$

and $w$ found from posing $w=k_1y+k_2$ and applying the boundary conditions

$
w=300\quad \text{when} \quad y=0 \quad \implies \quad k_2=300,
$

$
w=180\quad \text{when} \quad y=30/\sin60=2-\sqrt{3} \quad \implies \quad 180=20\sqrt{3k_1}+k_2  \quad \implies \quad  k_1=-2\sqrt{3},
$

so $w=300-2\sqrt{3}y$ (with units of metres), and

$
\displaystyle \delta F = \rho g \left(\frac{\sqrt{3}}{2}y\right)\left(300-2\sqrt{3}y\right)\delta y. \quad\dots(\ast)
$

Taking limits and integrating

$
\displaystyle F = \int_{y_1}^{y_2}\,dF = \frac{\sqrt{3}}{2} \rho g \int_{y_1}^{y_2} y\left(300-2\sqrt{3}y\right)\,dy = \frac{\sqrt{3}}{2} \rho g \left[300\frac{y^2}{2}-2\sqrt{3}\frac{y^3}{3}\right]_{0}^{20\sqrt{3}}
$

$
\implies \underline{F=1.12\,\textsf{GN}}.
$

To find the centre of pressure, take moments about OO, for an element

$
\delta M_{oo} = y\delta F.
$

Substituting for $\delta F$ from $(\ast)$, 

$
\delta M_{oo} = \rho g \frac{\sqrt{3}}{2}y^2\left(300-2\sqrt{3}y\right)\delta y.
$

Taking limits and integrating

$
\displaystyle M_{oo} = \frac{\sqrt{3}}{2} \rho g \int_{y_1}^{y_2} y^2\left(300-2\sqrt{3}y\right)\,dy = \frac{\sqrt{3}}{2} \rho g \left[300\frac{y^3}{3}-2\sqrt{3}\frac{y^4}{4}\right]_{0}^{20\sqrt{3}} 
$

$
\implies M_{oo}=24.72\,\textsf{GN}\,\textsf{m}.
$

But $M_{oo} = F_ny_p$, so

$
\displaystyle y_p = \frac{24.72}{1.12} = 22.07\,\textsf{m}
$

and 

$
\displaystyle h_p = \frac{\sqrt{3}}{2}y_p \quad \implies \underline{h_p=19.1\,\textsf{m}.}
$

::::
<!-- 
:::{exercise}📝
A rectangular plate is attached to a shaft along its centreline to form a butterfly valve as shown in Figure Q16. The valve is located in a $2\,\textsf{m}$ square duct. 

Find

1. The net hydrostatic force on the plate
2. The moment of that force about the shaft axis
3. The torque required to start to open the valve.

```{figure} images/rectangular_plate.png
:alt: Plate butterfly valve
:class: bg-none
:width: 70%
:align: center
```
