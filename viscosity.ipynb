# Viscosity

Let's revisit our concept of {ref}`fluid deformation<subsection_fluid_deformation>`.

```{figure} images/visc_def.png
:alt: Deformation of a fluid under shear
:class: bg-none
:width: 70%
:align: center
```

When a fluid is subjected to a shear stress, continuous deformation takes place as layers of fluid slide over one another. Viscosity describes the fluid's resistance to this motion, and is often described as 'internal friction'.

The higher the viscosity, the greater the shear stress required to sustain a given rate of deformation. Equivalently, for a given applied shear stress, a more viscous fluid deforms at a lower rate.

<div style="display: flex; gap: 20px; justify-content: center; align-items: flex-start;">

  <div style="width: 40%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/water_pouring.jpg"
         style="max-width: 100%; max-height: 100%;"
         alt="Water pouring">
  </div>

  <div style="width: 40%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/honey.jpeg"
         style="max-width: 100%; max-height: 80%;"
         alt="Honey">
  </div>

</div>


- Low viscosity (water) readily flows. 
- High viscosity (honey) resists flowing.

:::{dropdown} Does viscosity behave differently for liquids and gases?

Yes. The physical mechanisms responsible for viscosity differ between liquids and gases.

Liquids: Intermolecular forces contribute to the transfer of momentum between neighbouring layers moving at different velocities.

Gases: Molecules move randomly between layers with different mean velocities, transporting momentum as they move. Molecules moving from faster layers to slower ones tend to accelerate them, while molecules moving from slower layers to faster ones tend to retard them.

:::

:::{hint} Concept: Visualising viscosity
:class: concept

Viscosity can be thought of as a mechanism for the **diffusion of momentum** through a fluid. Momentum is transported between neighbouring layers, tending to reduce differences in their velocities.
:::

:::{tip} Definition: Boundary layer
:class: definition

At a solid surface, the fluid velocity is equal to the surface velocity. For a stationary surface, the fluid velocity is therefore zero.

There must be a region connecting no flow in contact with the surface to the main stream flow i.e. a velocity profile above the surface.

A **boundary layer** is this connecting region.

:::

Consider two adjacent layers at $y$ and $y+\delta y$ where the velocity is $u$ and $u + \delta u$ respectively. In the limit, as adjacent layers get infinitesimally close, $\delta y \rightarrow 0$ and the velocity gradient is $\displaystyle \dfrac{\text{d}u}{\text{d}y}$.


```{figure} images/bl_profilea.png
:alt: Velocity gradient in a boundary layer
:class: bg-none
:width: 60%
:align: center
```

:::{tip} Definition: Newton's law of viscosity
:class: definition

For a **Newtonian fluid**, the shear stress, $\tau$, is directly proportional to this velocity gradient perpendicular to the direction of flow

$
\tau = \mu \dfrac{\text{d}u}{\text{d}y}.
$

The constant of proportionality $\mu$ is the **dynamic viscosity** of the fluid; a measure of the fluid's resistance to shearing.
:::


| Symbol | Quantity | SI unit |
|:---:|:---|:---:|
| $\tau=F/A$ | Shear stress | $\mathsf{Pa}$ |
| $\mu$ | Dynamic viscosity | $\text{Pa}\,\text{s}$ |
| $\nu = \mu/\rho$ | Kinematic viscosity | $\text{m}^2/\text{s}$ |


:::{hint} Notes on viscosity
- The dynamic viscosity $\mu$ is often just referred to as the viscosity of the fluid. This is a fluid property.
- For liquids $\mu$ decreases with increasing temperature (reduced cohesive forces).
- For gases $\mu$ increases with increasing temperature (increased molecular activity).
- The **kinematic viscosity** $\nu= \mu/\rho$ is also a fluid property. Note the units of kinematic viscosity are those of a diffusion coefficient - highlighting viscosity as a diffusion of momentum. 
- Shear stress is related to **shear force** via

$
\displaystyle \tau = \frac{F}{A}\quad\implies\quad F=\tau A
$

The area is the surface over which the shear stress is acting – for a flat plate being dragged through water it would be the total surface area i.e. the top surface area plus the bottom surface area. 

:::

:::{dropdown} Do all fluids obey Newton's law of viscosity?

Most fluids are **newtonian** and with velocity gradient proportional to shear rate, but not all.

:::

:::{hint}Concept: Ideal fluid
:class: concept

An **ideal fluid** is considered **inviscid** and {ref}`incompressible<concept_compressibility>`.

Inviscid: Viscosity is neglected, so the fluid offers no resistance to shear deformation.

Incompressible: Density is constant; changes in pressure do not produce significant changes in volume. This implies no phase change occurs and the fluid remains liquid or gas throughout.

We also assume that surface tension is negligible.

The converse of an ideal fluid is a **real** fluid.

:::

:::{dropdown} Engineering example

```{figure} images/journal_bearing.png
:alt: Automotive journal bearing 
:class: bg-none
:width: 80%
:align: center
```

An example of lubrication theory in use is for the journal bearing of a car crankshaft. Here, the flow speed in the oil layer is given by the rotation rate $\omega$, by $u=r\omega$. Recall power is related to torque $T$ by $P=T\omega$, and torque is the product of force and radius $T = Fr$. 

Newton's law of viscosity gives us

$
\displaystyle \tau = \mu\frac{r\omega}{t}
$

with $t$ the thickness of the oil layer. The area over which the retarding torque is applied $A = 2\pi r L$. Therefore

$
\displaystyle T = \tau(2\pi r L)r = 2\tau\pi r^2L = 2\pi r^2L \mu\frac{r\omega}{t}=   2\pi r^3L \mu\frac{\omega}{t}
$

and the power required to overcome that retarding torque

$
\displaystyle P=  2\pi r^3L \mu\frac{\omega^2}{t}.
$


::::{prf:example} Hydraulic ram
:enumerator: 14

A hydraulic ram $200\,\text{mm}$ in diameter and $1.2\,\text{m}$ long moves wholly within a concentric cylinder $200.2\,\text{mm}$ in diameter, and the annular clearance is filled with oil of relative density $0.85$ and **kinematic viscosity** $400\,\text{mm}^2/\text{s}$.  What is the viscous force resisting the motion when the ram moves at $120\,\text{mm}/\text{s}$?

```{figure} images/hyd_ram.jpg
:alt: Hydraulic ram
:class: bg-none
:width: 30%
:align: center
```

:::{admonition} Solution
:class: dropdown

```{figure} images/ram.png
:alt: Hydraulic ram
:class: bg-none
:width: 70%
:align: center
```

The viscous force is related to the applied shear stress. For a Newtonian oil, noting that the kinematic viscosity $\nu = \mu/\rho$

$
\displaystyle \tau = \frac{F}{A} = \mu \dfrac{\text{d}u}{\text{d}y}= \nu \rho \dfrac{\text{d}u}{\text{d}y}.
$

But the gap between the ram and outer casing is thin with layer thickness\
$y=(d_o-d_i)/2$, so we can use the lubrication approximation

$
 \displaystyle \dfrac{\text{d}u}{\text{d}y} \approx \frac{u}{y}.
$

Therefore

$
 \displaystyle \tau = 2 \nu \rho \frac{u}{(d_o-d_i)}.
$

To translate to a resisting force on the ram, we need to understand what surface area the viscosity is acting over; which is the surface area of the ram

$
A = \pi d_i L.
$

$
 \displaystyle F = \tau A = 2 \nu \rho \frac{u}{(d_o-d_i)}  \pi d_i L
$

$
 \displaystyle F= 2\times 400\times 10^{-6} \times 1000 \times 0.85 \frac{120\times10^{-3}}{1\times10^{-3}}  \pi 200\times10^{-3}\times 1.2
$

$
\implies\underline{F = 307.6\,\text{N}}.
$

(section-Re)=
### Laminar and turbulent flows

*`Big whirls have little whirls*\
*Which feed on their velocity,* \
*And little whirls have lesser whirls*\
*And so on to viscosity.'*

Richardson, L.F., 1922. Weather Prediction by Numerical Process

```{figure} images/wake.png
:alt: Turbulent cloud wake
:class: bg-none
:width: 60%
:align: center
```

:::{hint} Concept: Laminar and turbulent flow
:class: concept

At low flow speeds, elements of fluid move smoothly, in straight lines.  The fluid can be considered as moving in layers or laminae. \
This is called **laminar flow**.

- [Laminar flow](https://www.youtube.com/watch?v=KqqtOb30jWs) in a pipe taken by B Carlisle and S Beck at Sheffield University (opens in YouTube)

At higher flow speeds, the paths of individual fluid particles are not straight, but disorderly, such that mixing takes place.\
This is called **turbulent flow**

- [Turbulent flow](https://www.youtube.com/watch?v=NplrDarMDF8) in a pipe taken by B Carlisle and S Beck at Sheffield University (opens in YouTube)

Between the fully-laminar and fully-turbulent flow is a **transitional flow**.



### Reynolds number

A turbulent flow contains eddies that carry momentum, drive mixing and accelerate neighbouring fluid. These eddies have **inertia**.

The forces due to this inertia ($F=ma$) are associated with
- The mass in the turbulent flow $m = \rho l^3$, where $l$ is a characteristic length scale. 
- Accelerations in the flow 

$
\displaystyle a = \dfrac{\text{d}u}{\text{d}t} \approx  \dfrac{u}{t} \propto \dfrac{u^2}{l},
$

where $u$ is a characteristic flow speed.

Hence, inertial forces are proportional to

$
\displaystyle \rho l^3 \dfrac{u^2}{l} = \rho u^2 l^2.
$

As flow slows down, viscosity damps these eddies, and viscous forces take over. These are proportional to

$
\displaystyle \mu\frac{u}{l}l^2 = \mu u l.
$


:::{tip} Definition: Reynolds number
:class: definition

The ratio of the inertial to viscous forces in a fluid flow is the **Reynolds number**, $\text{Re}$. 

$
\displaystyle \text{Reynolds number} = \frac{\text{inertial forces}}{\text{viscous forces}} = \frac{\rho u^2 l^2}{\mu u l}
$

$
\displaystyle \text{Re} = \frac{\rho u l}{\mu}
$

where density and viscosity ($\rho$ and $\mu$) are fluid properties, with $l$ and $u$ characteristic length and velocity scales within the flow. 

In laminar flow, viscous forces dominate and $\text{Re}$ is small.

In turbulent flow, inertia forces dominate and $\text{Re}$ is large.


:::

:::{dropdown} Non-dimensional groups

Reynolds number is a classic example of a non-dimensional group. It has no units because the units of the terms cancel out.

Non-dimensional groups are very important in fluid mechanics and need to be considered when scaling experiments etc. They tell us about the balance of forces or other dynamic competions within a flow. 

For example, consider flow in a circular section pipe.  If the Reynolds number is the same the behaviour of the fluid will be the same regardless of actual diameters, densities etc - because the inertia of the turbulent motion is matched to the viscous dissipation in the same balance. 

:::

### Characteristic scales? 

It is a little bit of an art to understand what 'characteristic' length and flow speed scales to take. But there are some standards. 

In **pipe flow** the characteristic length will be the pipe diameter $d$, and the characteristic speed will be the mean flow speed $v$. For a circular cross section pipe, the Reynolds number will be

$
\displaystyle \text{Re} = \frac{\rho v d}{\mu}
$

For flow in a **smooth circular pipe** under normal operating conditions, the following are typically assumed:

 - $\text{Re} < 2000$ : laminar flow
 - $2000 < \text{Re} < 4000$: transition
 - $\text{Re} > 4000$: fully turbulent flow

However, these figures can be significantly affected by surface roughness and background conditions. Flow in a rough pipe may be fully turbulent below $\text{Re} = 4000$. In contrast, Osborne Reynolds found that flow remained laminar to higher Reynolds numbers that flows today - often attributed to background instability from vibrations such as noise and traffic. 

For the flow over a **flat plate**

 - $\text{Re} < 5\times10^5$ : laminar flow
 - $5\times10^5 < \text{Re} < 3\times10^6$: transition
 - $\text{Re} > 3\times10^6$: fully turbulent flow.

Note, in this case, the boundary layer, where the flow speed approaches zero, will be laminar while the main free stream will be turbulent. The characteristic velocity scale for the Reynolds number is the free stream velocity, $u_\infty$, and the characteristic length scale is the distance from the edge of the plate, $x$

$
\displaystyle \text{Re} = \frac{\rho u_\infty x}{\mu}
$
