# Measuring pressure - manometry

Here we consider how the principles of hydrostatic pressure can be used in devices to measure pressure. 

### Piezometers

A **piezometer** is a simple device for measuring moderate pressures - or differences in pressure - in a liquid (but not a gas).

```{figure} images/piezometer.png
:alt: Piezometer schematic
:class: bg-none
:width: 80%
:align: center
```

The diagram above shows a pipeline containing liquid of density $\rho$. A piezometer is mounted to measure the pressure at point $A$. This consists of a thin, vertical, transparent tube with one end inserted into the top of the pipeline via a **tapping**, with the other end open to the atmosphere.

The elevated pressure in the pipeline liquid causes that liquid to rise up the piezometer tube to form a static column with height $h_A$. The free surface of this liquid column is at atmospheric pressure. 

Therefore, at the base of the liquid column the **gauge** pressure must be $p_A=\rho gh_A$.

Similarly, if a second piezometer tube is placed at tapping B, pressure in the liquid at $B$ causes a static column of pipeline liquid with height $h_B$ to form. The gauge pressure $p_B$ at point $B$ is $p_B=\rho gh_B$.

Together, the two piezometers measure the pressure difference $p_A-p_B$ between the two points

$
p_A-p_B=\rho gh_A-\rho gh_B = \rho g(h_A-h_B).
$

The pressure difference is determined by measuring the difference in height between the two liquid columns ($h_A-h_B$), given the known pipeline fluid density $\rho$. 

:::{hint} Notes on piezometers
- The measured pressures are gauge pressures. Because the piezometer is open to the atmosphere, the hydrostatic pressure due to the static column height will be relative to atmospheric.
- A piezometer is open to the atmosphere and so is only suitable to measure pressures in liquids and not gasses. A gas would simply escape. 

### Manometers

A manometer shares the underlying principle of the piezometer, with the addition of a second **liquid** between the pipeline fluid and the atmospheric free surface. This means that a manometer can measure pressures in any fluid - liquid or gas.

```{figure} images/manometer.png
:alt: Manometer schematic
:class: bg-none
:width: 90%
:align: center
```

The intermediary fluid is termed the **"working liquid" or "manometric liquid"**, with density $\rho_m$. In the configuration shown, with the working liquid sitting below the pipeline liquid, $\rho_m>\rho$; the two are immiscible, such that an interface forms between them.

The pressure in the pipeline forces some of the fluid into the manometer tube, pushing the working fluid, such that the positions of the interfaces in each limb of the manometer shift. The pressure $p_A$ at point $A$ is directly related to this shift in interface levels; to be found by balancing the hydrostatic contributions of the different fluids in the manometer. 

Drawing an imaginary horizontal line ($X$ in the diagram) through the lower interface, we note that the pressure in the left-hand limb at this level must be the same as the pressure in the right-hand limb. This must be true for the manometer to be in equilibrium - i.e. with no flow. 

From the left-hand limb, the pressure at the interface

$
p_X=p_A+\rho gh_1.
$

Similarly in the right-hand limb 
$
p_X=\rho_mgh_2.
$

Equating eliminates $p_X$ and

$
p_A=g(\rho_mh_2-\rho h_1).
$

Hence the pressure $p_A$ can be determined by measuring the two heights $h_1$ and $h_2$, for the known fluid densities $\rho$ and $\rho_m$.

For example, if the pipeline above contains air with density $\rho=1.2\,$kg/m$^3$ and the working fluid is water with density $\rho_m=1000\,$kg/m$^3$, with the heights measured $h_1=0.3\,$m and $h_2=0.1\,$m, we can evaluate the gauge pressure in the air

$
p_A=g(\rho_mh_2-\rho h_1)=9.81(1000\times0.1-1.2\times0.3)\,$Pa

$
\implies \underline {p_A=977\,\text{Pa}}.
$

:::{hint} Notes on manometers
- The pressures above are all gauge pressures. The tube is open to the atmosphere which creates the reference pressure.
- If the pipeline fluid is gas, the manometric liquid is typically water $\rho_m=\rho_{H_2O}=1000\,$kg/m$^3$ or some form of oil $\rho_m\approx 800$ to $900\,$kg/m$^3$.
- If the pipeline contains a flowing liquid, the manometric liquid is typically mercury $\rho_m=\rho_{Hg}=13600\,$kg/m$^3$.
- If we have no reference pressure, we can still evaluate pressure differences using this method using a differential manometer (see worked example).
:::

:::{tip} Evaluating pressures from a manometer
:class: definition
1. Pressures are constant in a horizontal plane: identify the plane that exists in two arms of the manometer.
2. Construct the hydrostatic balances in each limb, recalling that the {ref}`hydrostatic pressure <def-hydrostatic>` at the base of a water column is that at the top of the column, plus the hydrostatic contribution $\rho g h$ of the column. 
3. Apply equilibrium by equating the pressures in the two limbs at the identified plane. 

### Differential manometer

:::{prf:example} Differential manometer
:enumerator: 7

Two pipelines convey a gas with density $\rho$. A **differential manometer** with a working fluid of density $\rho_m$ measures the pressure difference $p_A-p_B$, between point A and point B in the respective pipelines. 

```{figure} images/diff_man.png
:alt: Differential manometer schematic
:class: bg-none
:width: 80%
:align: center
```
The pressure difference can be found by measuring the heights $h_1$ and $h_2$. Show that

$
p_A-p_B=(\rho_m-\rho)(h_1-h_2)g,
$

and hence find the pressure difference when $\rho=1.2\,$kg/m$^3$, $\rho_m=800\,$kg/m$^3$, $h_1=58\,$mm and $h_2=176\,$mm.

::::{admonition} Solution
:class: dropdown

We note that the horizontal plane $X$ intersects each limb of the manometer and exists within one fluid; applying equilibrium at $X$ in each limb will relate $p_A$ and $p_B$. 

In the left-hand limb

$
p_X=p_A+\rho gh_1 + \rho_m g(h_2-h_1),
$

and in the right-hand limb

$
p_X=p_B+\rho g h_2.
$

Equating eliminates $p_X$, with 

$
p_A-p_B= \rho g h_2 - \rho gh_1 - \rho_m g(h_2-h_1).
$

Collecting terms in $\rho$ and $\rho_m$, we can rearrange

$
p_A-p_B= \rho g(h_2-h_1) - \rho_m g(h_2-h_1) = g(\rho -\rho_m)(h_2-h_1)
$

or

$
\underline{p_A-p_B= g(\rho_m -\rho)(h_1-h_2)}
$

as required. 

:::{dropdown} Evaluating

Evaluating from the data provided

$
p_A-p_B= g(\rho_m -\rho)(h_1-h_2) = 9.81\times(800-1.2)(58-176)\times 10^{-3}$\,Pa,

$
\implies \underline{p_A-p_B=-925\,\mathsf{kPa}.}
$

:::{prf:example} 🔥 Challenge manometer example 
:enumerator: 8

A manometer consists of two tubes A and B, with vertical axes and uniform cross-sectional areas $A_a=500\,$mm$^2$ and $A_b=800\,$mm$^2$ respectively, connected by a U-tube C of cross-sectional area $A_c=70\,$mm$^2$ throughout - shown in the figure below.  Tube A contains a liquid of relative density $\Rho_a=0.8$, tube B contains one of relative density $\Rho_b=0.9$.  The surface of separation between the two liquids is in the vertical side of C connected to tube A.  Determine the additional pressure, $\Delta p_B$ that, when applied to tube B, will cause the surface of separation to rise $\Delta h_c=60\,$mm in tube C.

```{figure} images/manometer_challenge.png
:alt: Manometer complex schematic
:class: bg-none
:width: 70%
:align: center
```
::::{admonition} Solution
:class: dropdown

The surface in tube B moves down a distance $\Delta h_b$ due to the additional pressure applied, and liquid B is pushed into tube C. The interface between the two fluids in the left hand manometer arm ("the surface of separation") is pushed upwards due to this additional fluid.

The rise in the surface of separation will push liquid A up into tube A, raising the surface in tube A a distance $\Delta h_a$ - shown in the diagram below. 

```{figure} images/manometer_challenge_soln.png
:alt: Manometer complex schematic
:class: bg-none
:width: 70%
:align: center
```
1. The volume change in tubes A, B and C must match each other. We can express this mathematically in terms of the height changes and the constant cross_sectional areas

$
A_c\Delta h_c = A_b\Delta h_b = A_a\Delta h_a.
$

From here, we can find the changes in depth in each tube

$
\displaystyle \Delta h_a = \frac{A_c\Delta h_c}{A_a}, \quad \text{and} \quad \Delta h_b = \frac{A_c\Delta h_c}{A_b}.
$

2. Before the extra pressure was added in tube B, the manometer was in equilibrium with the fluid depths in tubes A and B $h_a$ and $h_b$ respectively. In the left-hand arm at level 1 (horizontal plane of constant pressure existing in both arms) in the diagram

$
p_1=p_A+\rho_a gh_a
$

and in the right-hand arm

$
p_1=p_B+\rho_b gh_b.
$

3. Equilibrium in the manometer after the pressure is added is given at level 2 in the diagram. In the left-hand arm 

$
p_2=p_A+\rho_a g(h_a+\Delta h_a - \Delta h_c),
$

and in the right-hand arm

$
p_2=p_B+\Delta p_B + \rho_b g(h_b-\Delta h_b - \Delta h_c).
$

4. Subracting the before from the after scenario in each arm can give us a pair of equations for $p_2-p_1$. 

In the left-hand arm

$
p_2-p_1 = p_A+\rho_a g(h_a+\Delta h_a - \Delta h_c) - p_A -\rho_a gh_a,
$

which simplifies to
$
p_2-p_1=\rho_a g(\Delta h_a - \Delta h_c).
$

Similarly in the right-hand arm

$
p_2-p_1 = p_B+\Delta p_B + \rho_b g(h_b-\Delta h_b - \Delta h_c) - p_B - \rho_b gh_b,
$

simplifying to

$
p_2-p_1 = \Delta p_B - \rho_b g(\Delta h_b + \Delta h_c).
$

Equating $p_2-p_1$ from the left and right-hand arms

$
\rho_a g(\Delta h_a - \Delta h_c)= \Delta p_B - \rho_b g(\Delta h_b + \Delta h_c),
$

which can be rewritten for the pressure change we are looking for

$
\Delta p_B = \rho_a g(\Delta h_a - \Delta h_c) + \rho_b g(\Delta h_b + \Delta h_c).
$

5. From our first step, we related the changes in depth in each tube, so we can write $\Delta p_B$ now in terms of data provided

$
\displaystyle \Delta p_B = g\rho_w\left\{\Rho_a\left[\Delta h_c\left( \frac{A_c}{A_a} - 1\right)\right] + \Rho_b\left[\Delta h_c\left( \frac{A_c}{A_b} + 1\right)\right]\right\}.
$

6. Evaluating from the data provided

$
\displaystyle \Delta p_B = 9.81\times1000\left\{0.8\left[\frac{60}{1000}\left( \frac{70}{500} - 1\right)\right] + 0.9\left[\frac{60}{1000}\left( \frac{70}{800} + 1\right)\right]\right\}\,\mathsf{Pa},
$

$
\implies \underline{\Delta p_B = 172\,\mathsf{Pa}}.
$

::::{admonition} Your work

You can now completee the {ref}`Fluid Properties chapter exercise page <section-fluid_properties_exercises>`.
