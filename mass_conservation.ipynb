# Mass conservation - Continuity

### Terminology 

Now we need the vocabulary to describe quantities in flow. 

:::{tip} Definition: Volume flow rate, Mass flow rate
:class: definition

```{figure} images/volume_flow_rate.png
:alt: Representation of uniform flow in a pipe
:class: bg-none
:width: 80%
:align: center
```

Consider the pipe section above, of cross-sectional area $A$, containing fluid of density $\rho$, flowing with velocity $u$.

In a small time interval $\Delta t$ the volume of fluid, $\Delta V$, and mass of fluid, $\Delta m$, that flows through area $A$
are given by

$
\Delta V = Au\Delta t\quad \text{and}\quad \Delta m = \rho \Delta V = \rho Au\Delta t.
$

The volume flow rate is defined

$
\displaystyle Q = \frac{\Delta V}{\Delta t} = Au,
$

which is the volume of fluid which passes through a given cross-sectional area in a given time.

Similarly mass flow rate

$
\displaystyle \dot m = \frac{\Delta m}{\Delta t} = \rho Au = \rho Q,
$

which is the mass of fluid which passes through a given cross-sectional area in a given time.

:::


| Symbol | Quantity |  | SI unit |
|:---:|:---|:---:|:---:|
| $Q$ | Volume flow rate |  | $\text{m}^3/\text{s}$ |
| $\dot m$ | Mass flow rate| | $\text{kg}/\text{s}$ |


:::{tip} Definition: Control volume
:class: definition

A **control volume** (CV) is a fixed region through which fluid flows.

Budgets of conserved quantities over a control volume allow us to derive equations describing variation of those quantities through a flow field.

We consider transfers across the boundary of the control volume and changes within the volume inside the boundary.

The boundary of control volume is chosen to link places where we know the fluid properties (i.e. velocity, mass flow rate etc) to places we want to find out the flow properties

:::

<div style="display: flex; gap: 20px; justify-content: center; align-items: flex-start;">

  <div style="width: 30%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/cv1.png"
         style="max-width: 100%; max-height: 100%;"
         alt="Control volume choice example">
  </div>

  <div style="width: 60%; height: 240px; display: flex; justify-content: center; align-items: center;">
    <img src="images/cv2.png"
         style="max-width: 100%; max-height: 100%;"
         alt="Control volume choice example">
  </div>

</div>


In the absence of nuclear reactions, matter is neither created not destroyed. We can apply this principle of mass conservation to a flowing fluid.

:::{hint} Concept: Conservation of mass
:class: concept

The principle of mass conservation states that, for any control volume in any flow

<table style="width:85%; table-layout:flex;">
  <colgroup>
    <col style="width:30%;">
    <col style="width:3%;">
    <col style="width:30%;">
    <col style="width:3%;">
    <col style="width:30%;">
  </colgroup>
  <tr>
    <th>mass flow rate entering
the control volume</th>
    <th>=</th>
    <th>mass flow rate leaving the
control volume</th>
    <th> + </th>
    <th>rate of change of mass inside the control volume</th>
  </tr>
</table>

$
\displaystyle \dot m_{\text{in}} = \dot m_{\text{out}} + \dfrac{\text{d}m}{\text{d}t}
$

:::

:::{hint} Notes on mass conservation - continuity

- Most of the flows we consider are steady, in which case 

$
\displaystyle \dfrac{\text{d}m}{\text{d}t} = 0\quad \implies \quad \dot m_{\text{in}} = \dot m_{\text{out}}.
$

- When we consider incompressible, homogeneous fluids $\rho =$ constant. Noting that $m = \rho V$, $\dot m = \rho Q$, and conservation of mass reduces to the **continuity equations**
    - for unsteady flow $\displaystyle Q_{\text{in}} = Q_{\text{out}} + \dfrac{\text{d}V}{\text{d}t}$
    - for steady flow $Q_{\text{in}} = Q_{\text{out}}$

where $\displaystyle \dfrac{\text{d}V}{\text{d}t}$ is the rate of change of the fluid volume inside the control volume.

In this case mass conservation reduces to volume conservation, also known as **continuity**.

:::

::::{prf:example} Steady flow through a constriction
:enumerator: 16

Consider the steady flow of an ideal fliuid through a constricted pipe.

```{figure} images/constriction.png
:alt: Steady flow through a constriction
:class: bg-none
:width: 80%
:align: center
```

Find $u_{\text{out}}$ if $d_{\text{in}}=10\,\text{cm}$, $u_{\text{in}}=0.5\,\text{m}/\text{s}$ and $d_{\text{out}}=3\,\text{cm}$.

:::{admonition} Solution
:class: dropdown

The flow is ideal i.e. inviscid and incompressible, with constant $\rho$. Therefore mass conservation reduces to continuity, over the control volume shown 

$
Q_{\text{in}} = Q_{\text{out}}\quad \implies \quad u_{\text{in}}A{\text{in}} = u_{\text{out}}A{\text{out}}.
$

In terms of pipe diameters

$
\displaystyle u_{\text{in}}\frac{\pi d_\text{in}^2}{4} = u_{\text{out}}\frac{\pi d_\text{out}^2}{4},
$

which cancelling $\pi/4$s and rearranging

$
\displaystyle  u_{\text{out}} = u_{\text{in}}\frac{d_\text{in}^2}{d_\text{out}^2} = 0.5\times \frac{10^2}{3^2}
$

$
\implies \underline{u_{\text{out}} = 5.56\text{m}/\text{s}.}
$



:::{hint}

Note: The flow speed in a pipe varies with the square of the diameter.

:::

::::{prf:example} Relating to flow speeds
:enumerator: 17

A $50\,\text{litre}$ tank takes $4\,\text{minutes}$ to fill with petrol (density $733\,\text{kg}/\text{m}^3$) through a nozzle of diameter $15\,\text{mm}$.  

1. What is the mass flowrate of petrol into the tank?  
2. What is the mean velocity of petrol leaving the nozzle?


:::{admonition} Solution
:class: dropdown

1. The mass flow rate by definition

$
\dot m = \rho Q
$

where $Q$ is the volume flow rate entering the petrol tank, which is the volume entering the tank per unit time. Therefore

$
\displaystyle \dot m = 733 \times \frac{50\times10^{-3}}{4\times 60}\,\frac{\text{kg\,\text{min}}\,\text{m}^3}{\text{m}^3\,\text{min}\,\text{s}}
$

$
\implies \underline{ \dot m = 0.153\,\text{kg}/\text{s}.}
$

2. The mean velocity in the nozzle can be found because the mass flow rate exiting the nozzle is equal to the mass flow rate entering the tank

$
\displaystyle \dot m = \rho u A \quad \implies \quad u = \frac{\dot m}{\rho A} = \frac{0.153\times 4}{733\times\pi(15\times10^{-3})^2}
$


$
\implies \underline{u = 1.18\,\text{m}/\text{s}.}
$
