The floating cube is in equilibrium, therefore it's weight balances the upward buoyancy force, which is the weight of the displaced fluid volume

$
F_B=h_il^2\rho_w g = \rho l^3 g.
$

Rearranging for the cube density $\rho$

$
\displaystyle \rho = \frac{h_i\rho_w}{l} = \frac{27\times 1000}{35}\,\frac{\textsf{mm}\,\textsf{kg}}{\textsf{mm}\,\textsf{m}^3}
$

$
\implies \underline{\rho = 771\,\textsf{kg}/\textsf{m}^3}.
$



2. 

```{figure} images/we9b.png
:alt: New depth of immersion for floating cube with scoop removed
:class: bg-none
:width: 60%
:align: center
```

Now the weight of the cube is changed, because some material is removed. The new equilibrium is given by 

$
\displaystyle F_B=hl^2\rho_w g = \rho g \left(l^3 - \frac{1}{2}\frac{\pi d^3}{6}\right).
$

Rearranging for the new immersion depth 

$
\displaystyle h = \frac{\rho}{l^2\rho_w} \left(l^3 -\frac{\pi d^3}{12}\right) = \frac{771}{35^2\times 1000} \left(35^3 -\frac{\pi 15^3}{12}\right)
$


$
\implies \underline{h = 26.4\,\textsf{mm}}.
$

::::{prf:example} Helium balloon
:enumerator: 10

A helium balloon has a mass of $300\,\textsf{kg}$ (balloonist, balloon structure etc) with an additional $400\,\textsf{kg}$ of sand ballast is also carried.

```{figure} images/met_balloon_qu.png
:alt: Meteorological balloon being filled with helium gas
:class: bg-none
:width: 60%
:align: center
```

1. At ground level the air pressure is $1.01\,\textsf{bar}$ and the temperature is $300\,\textsf{K}$. Calculate the volume of helium that must be put into the balloon so that it can rise.

($\tilde R$ for air is $287.1\,\textsf{J}/\textsf{kg}\,\textsf{K}$; $\tilde R$ for helium is $2077\,\textsf{J}/\textsf{kg}\,\textsf{K}$).

2. The balloon rises to $2000\,\textsf{m}$ where the air pressure is $0.795\,\textsf{bar}$ and the temperature is $275\,\textsf{K}$. To what volume will the balloon expand?

3. The balloon is now completely full and cannot expand any more. Helium must be vented if it rises any higher to prevent it bursting. The balloon needs to rise further to get over some mountains. How much sand ballast must be released if the balloon is to rise to $3000\,\textsf{m}$ where the air pressure is $0.7012\,\textsf{bar}$ and the temperature is $269\,\textsf{K}$?

4. Show that the buoyancy of the balloon is related only to the mass of helium it is filled with, provided that the balloon is in equilibrium with the surrounding air.

:::{admonition} Solution
:class: dropdown

1. The total mass of the balloon plus ballast plus additional mass of helium added must be balanced by the buoyancy of the added helium which is the weight of air displaced

$
m_bg+V_b\rho_{He}g = V_b\rho_{\text{air}}g,
$

$
m_b = V_b(\rho_{\text{air}}-\rho_{He}).
$

From the Perfect Gas Law, we can relate the densities to the gas constants provided in the question, since 

$
\displaystyle p=\rho\tilde R T \quad\implies\quad \rho =\frac{p}{\tilde R T}.
$

Therefore 

$
\displaystyle m_b=\frac{pV_b}{T}\left(\frac{1}{\tilde R_{\text{air}}}-\frac{1}{\tilde R_{He}}\right)\quad\dots(\dagger)
$

and

$
\displaystyle V_b=\frac{Tm_b}{p}\left(\frac{1}{\tilde R_{\text{air}}}-\frac{1}{\tilde R_{He}}\right)^{-1}\quad\dots(\ast)
$


At ground level $p=1.01\,\textsf{bar}$ and the temperature is $300\,\textsf{K}$; evaluating for the volume of helium to be added at ground level for the balloon to rise

$
\displaystyle V_b=\frac{300\times 700}{1.01\times 10^5}\left(\frac{1}{287.1}-\frac{1}{2077}\right)^{-1}
$

$
\implies\underline{V_b=692\,\textsf{m}^3}.
$

2. The balloon rises to $2000\,\textsf{m}$ where the air pressure is $0.795\,\textsf{bar}$ and temperature $275\,\textsf{K}$ (assuming the balloon is in equilibrium with the air). From equation ($\ast$)

$
\displaystyle V_b=\frac{275\times 700}{0.795\times 10^5}\left(\frac{1}{287.1}-\frac{1}{2077}\right)^{-1}
$

$
\implies\underline{V_b=806.6\,\textsf{m}^3},
$

at $2000\,\textsf{m}$.

3. The balloon rises further, to $3000\,\textsf{m}$, where the air pressure is $0.7012\,\textsf{bar}$ and temperature $269\,\textsf{K}$, but the volume remains at $V_b=806.6\,\textsf{m}^3$. Sand ballast must be ditched. From ($\dagger$), the target mass of the balloon will be 

$
m_b = \frac{0.7012\times10^5\times 806.6}{269}\left(\frac{1}{287.1}-\frac{1}{2077}\right)
$

$
\implies m_b=631.2\,\textsf{kg}.
$

So the balloonist must jettison $\underline{700-631.2=68.8\,\textsf{kg}}$ of sand ballast. 

4. The Perfect Gas Law for helium states

$
\displaystyle \frac{pV_b}{T} = m_{He}\tilde R_{He}
$

(the helium has displaced any air in the balloon). Combining with equation ($\dagger$)

$
\displaystyle m_b=\frac{pV_b}{T}\left(\frac{1}{\tilde R_{\text{air}}}-\frac{1}{\tilde R_{He}}\right)
$

$
\displaystyle m_b=m_{He}\left(\frac{\tilde R_{He}}{\tilde R_{\text{air}}}-1\right).
$

Therefore the mass that can be lifted $m_b$ depends only on the mass of helium added.
