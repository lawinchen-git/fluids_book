### Density

:::{exercise} 💧

Calculate the density of compressed air stored in a $0.1\,\mathsf{m}^3$ canister with total mass total mass $25.2\,\mathsf{kg}$. The canister itself has mass $26.1\,\mathsf{kg}$.

::::{dropdown} ✓ Check your answer
$9\,\mathsf{kg}/\mathsf{m}^3$
::::

::::{admonition} Solution
:class: dropdown
Here, we have the mass of air $m=(26.1-25.2)\,\mathsf{kg}=0.9\,\mathsf{kg}$ occupying $V=0.1\,\mathsf{m}^3$. Hence,

$\displaystyle\rho=\frac{26.1-25.2}{0.1}\,\mathsf{kg}/\mathsf{m}^3$   

$\implies \underline{\rho = 9\,\mathsf{kg}/\mathsf{m}^3}$.

::::

:::{exercise} 💧
A $0.005\,\mathsf{m}^3$ sample of hydraulic oil has density $860\,\mathsf{kg}/\mathsf{m}^3$ at $15^\circ$C. Calculate the mass of the sample.

::::{dropdown} ✓ Check your answer
$4.3\,\mathsf{kg}$
::::


::::{admonition} Solution
:class: dropdown
A sample with density $\rho = 860\,\mathsf{kg}/\mathsf{m}^3$ and volume $V=0.005\,\mathsf{m}^3$ has mass $m=\rho V$. Hence,

$m = 860\times 0.005\,\mathsf{kg}\,\mathsf{m}^3/\mathsf{m}^3$

$
\implies \underline{m = 4.3\,\mathsf{kg}}.
$

:::{exercise}💧
If $4\,\mathsf{m}^3$ of oil weighs $35\,\mathsf{kN}$, calculate its density and relative density.

::::{dropdown} ✓ Check your answer
$892\,\mathsf{kg}/\mathsf{m}^3, \quad 0.892$
::::

::::{admonition} Solution
:class: dropdown
Here, we have the weight of oil $mg=35\times 10^3\,\mathsf{N}$ giving a mass $m=35\times 10^3/9.81\,\mathsf{N s}^2/\mathsf{m}$. This occupies a volume $V=4\,\mathsf{m}^3$. Hence,

$
\displaystyle\rho=\frac{35\times10^3}{9.81\times4}\,\mathsf{N s}^2/\mathsf{m}^4  
$

$
\implies \underline{\rho = 892\,\mathsf{kg}/\mathsf{m}^3},
$
and the relative density - i.e.~the ratio of density to the density of water, is

$
\displaystyle\Rho = \frac{\rho}{\rho_w}=\frac{892}{1000}\,\mathsf{kg}/\mathsf{m}^3\,\mathsf{m}^3/\mathsf{kg}
$
$
\implies \underline{\Rho = 0.892}.
$

The volume of air $V=\mathcal{v} m$, where specific volume $\mathcal{v}=0.60\,\mathsf{m}^3/\mathsf{kg}$ and mass $m=0.50\,\mathsf{kg}$. Hence,

$
V=0.6 \times 0.5\,\mathsf{m}^3 \mathsf{kg}/\mathsf{kg}
$
$
\implies \underline{V = 0.3\,\mathsf{m}^3}.
$

:::{exercise} 💧
A gas in a storage vessel occupies $15\,\mathsf{m}^3$ and has a **specific volume** of $0.75\,\mathsf{m}^3/\mathsf{kg}$. Calculate the mass of gas in the vessel.

::::{dropdown} ✓ Check your answer
$20\,\mathsf{kg}$
::::

::::{admonition} Solution
:class: dropdown
The mass $m$ of the gas with specific volume $\mathcal{v} = 0.75\,\mathsf{m}^3/\mathsf{kg}$ and volume $V=15\,\mathsf{m}^3$ is $m=V/\mathcal{v}$. Hence,

$\displaystyle m=\frac{15}{0.75}\,\mathsf{m}^3 \mathsf{kg}/\mathsf{m}^3$

$\implies \underline{m = 20\,\mathsf{kg}}$.

Calculate the total mass of liquid and air contained within the spray cloud.

::::{dropdown} ✓ Check your answer

::::

::::{admonition} Solution
:class: dropdown
The bulk density of the spray $\rho=m/V = 25\,\mathsf{kg}/\mathsf{m}^3$ with volume $V=0.40\,\mathsf{m}^3$. Hence, the total mass of liquid and air $m = \rho V$ with
$
m= 25 \times 0.40\,\mathsf{kg} \mathsf{m}^3/\mathsf{m}^3  
$
$

### Perfect Gas Law

:::{exercise}📝
A piston-cylinder device contains air at an absolute pressure of $250\,\mathrm{kPa}$ and a temperature of $20^\circ\mathrm{C}$. The air occupies a volume of $0.080\,\mathrm{m^3}$.

The piston is allowed to move freely, and the air expands until its pressure is reduced to $150\,\mathrm{kPa}$. During the expansion, the air cools to a temperature of $5^\circ\mathrm{C}$.

Assuming that the air behaves as a perfect gas, determine the final volume of the air in $\mathrm{m^3}$.

::::{dropdown} ✓ Check your answer
$0.126\,\mathrm{m^3}$
::::

::::{admonition} Solution
:class: dropdown

For a fixed mass of perfect gas,

$\displaystyle \frac{p_1V_1}{T_1}=\frac{p_2V_2}{T_2}$,

which rearranges for final volume

$\displaystyle V_2=\frac{p_1V_1T_2}{p_2T_1}$.


Converting temperatures to Kelvin,

$
T_1 = 20+273.15 = 293.15\,\mathrm{K} \quad\mathrm{and}\quad T_2 = 5+273.15 = 278.15\,\mathrm{K},
$

we can substitute values

$
\displaystyle V_2=\frac{(250\times10^3\times 0.080\times278.15}{150\times10^3\times 293.15}.
$

$
\implies \underline{V_2 = 0.126\,\mathrm{m^3}}
$

The air therefore expands from $0.080\,\mathrm{m^3}$ to approximately $0.126\,\mathrm{m^3}$.

### Pressure

:::{exercise}💧

Find the pressure relative to atmospheric pressure experienced by a diver working on the sea bed at a depth of $35\,\mathsf{m}$. Take the density of sea water to be $1030\,\mathsf{kg}/\mathsf{m}^3$.

::::{dropdown} ✓ Check your answer
$353.7\,\mathsf{kPa}\approx 3.54\,\mathsf{bar}$
::::

::::{admonition} Solution
:class: dropdown

The pressure relative to atmospheric pressure is the gauge pressure due to the liquid above the diver,

$
p=\rho g h
$

where $\rho=1030\,\mathsf{kg}/\mathsf{m}^3$, $g=9.81\,\mathsf{m}/\mathsf{s}^2$ and $h=35\,\mathsf{m}$. Hence,

$
p=1030\times9.81\times35\,\mathsf{Pa}
$

$
\implies \underline{p=353.7\,\mathsf{kPa}\approx3.54\,\mathsf{bar}}.
$

:::{hint} 
To check think through your units, with pressure $p=F/A \implies \mathsf{N}/\mathsf{m}^2$, force $F=ma \implies \mathsf{kg}\,\mathsf{m}/\mathsf{s}^2$ or $\mathsf{N}$. Therefore pressure has units $\mathsf{kg}\,\mathsf{m}/\mathsf{s}^2\mathsf{m}^2$, simplifying to $\mathsf{kg}/\mathsf{m}\,\mathsf{s}^2$ which is Pascals $\mathsf{Pa}$.
:::

:::{exercise}💧

An open glass is sitting on a table. It has a diameter of $10\,\mathsf{cm}$. If water up to a height of $20\,\mathsf{cm}$ is now added, calculate the force exerted onto the table by the addition of the water.

::::{dropdown} ✓ Check your answer
$15.41\,\mathsf{N}$
::::

::::{admonition} Solution
:class: dropdown

You can ignore both the weight of the glass and atmospheric pressure since they were there originally and you are only interested in the extra force exerted onto the table due to the addition of the water.

The force exerted on the table by the water is equal to the weight dues to the hydrostatic pressure acting on the base of the glass, $p=\rho g h$. 

Force $F=pA$, where $A$ is the surface area of the glass base. 

Hence, 

$\displaystyle F=\rho g h \pi r^2 = 1000\times9.81\times\frac{20}{100}\pi\left(\frac{10}{2\times 100}\right)^2\,\frac{\mathsf{kg\,m\,m}\,\mathsf{m}^2}{\mathsf{m}^3\mathsf{s}^2}$

$ \implies \underline{F=15.41\,\mathsf{N}}$

:::{exercise} 📝

A closed rectangular tank with internal dimensions $6\,\mathsf{m}\times3\,\mathsf{m}\times1.5\,\mathsf{m}$ high has a vertical riser pipe of cross-sectional area $0.001\,\mathsf{m}^2$ in the upper surface, as shown in Figure Q10. The tank and riser are filled with water such that the water level in the riser pipe is $3.5,\mathsf{m}$ above the top of the tank. Atmospheric pressure is $1\,\mathsf{bar}$.

```{figure} images/riser.png
:alt: Schematic of a rectangular tank with riser
:class: bg-none
:width: 40%
:align: center
```
Figure Q10: Sketch of tank with riser (not to scale)

Calculate
1. The gauge pressure at the top of the tank.
2. The gauge pressure at the base of the tank.
3. The force exerted on the base of the tank due to the gauge water pressure.
4. The weight of water in the tank and riser.
5. Explain the difference between (3) and (4).
::::{hint} Hint
:class: dropdown
It may be helpful to think about the force on the top of the tank.
::::

::::{dropdown} ✓ Check your answer
1. $34.3\,\mathsf{kPa}$
2. $49.05\,\mathsf{kPa}$
3. $882.9\,\mathsf{kN}$
4. $264.9\,\mathsf{kN}$
::::

::::{admonition} Solution
:class: dropdown

The water level in the riser is $3.5\,\mathsf{m}$ above the top of the tank. Therefore, the pressure at the top of the tank is determined by a water column of height $3.5\,\mathsf{m}$.

1. The gauge pressure at the top

$ p_{\mathrm{top}}=3.5 \rho_w g $, $ \implies \underline{p_{\mathrm{top}}=34.3\,\mathsf{kPa}} $.

2. The total depth from the water surface in the riser to the base of the tank

$ h=3.5+1.5=5.0\,\mathsf{m}$.

Hence, the gauge pressure at the base

$ p_{\mathrm{base}}=5.0 \rho_w g $,  $ \implies \underline{p_{\mathrm{base}}=49.05\,\mathsf{kPa}} $

3. The area of the base of the tank

$ A=6\times3=18\,\mathsf{m}^2$.

Therefore, the force exerted on the base due to the gauge pressure

$ F_{\mathrm{base}}=p_{\mathrm{base}}A$ 

$ =49.05\times10^3\times18\,\mathsf{N}$, $ \implies \underline{F_{\mathrm{base}}=882.9\,\mathsf{kN}}$

4. The volume of water in the rectangular tank

$ V_{\mathrm{tank}}=6\times3\times1.5=27\,\mathsf{m}^3 $

and the volume of water in the riser

$ V_{\mathrm{riser}}=0.001\times3.5=0.0035\,\mathsf{m}^3$.

Thus,

$ V_{\mathrm{total}}=27.0035\,\mathsf{m}^3 $ 

and the weight of water

$ W=\rho_w V_{\mathrm{total}}g $, $ \implies \underline{W=264.9\,\mathsf{kN}} $.

5. The force on the base is therefore much greater than the weight of water. This is because the pressure at the top surface of the tank is also above atmospheric pressure, producing an upward force on the top of the tank. The difference between the downward force on the base and the upward force on the top is the weight of the water.


::::


### Manometry

:::{exercise} 📝

A double U-tube manometer is connected to a pipe as shown in Figure Q11. Taking the dimensions and fluids as indicated, calculate the absolute pressure at point A (centre of the pipe).

Take atmospheric pressure as $1.01\,\mathsf{bar}$ and the density of mercury as $13600\,\mathsf{kg}/\mathsf{m}^3$.

```{figure} images/double_u_man.png
:alt: Double U-tube manometer
:class: bg-none
:width: 60%
:align: center
```
Figure Q11: Double U-tube manometer with readings, for use with exercise 11 (not to scale)


::::{dropdown} ✓ Check your answer
$124.7\,\mathsf{kPa}$
::::

::::{admonition} Solution
:class: dropdown

We start the analysis from the point where we know what the condition is, in this case atmospheric pressure on the far right of the system. We then chose horizontal planes within the same fluid where there is an interface, this is because we know that the pressure is constant since pressure is constant at any horizontal plane in the same fluid. We chose interface points as this allows us to define each point in the system as a function of the known pressure, atmospheric pressure on the far right. 

```{figure} images/double_u_man_soln.png
:alt: Double U-tube manometer
:class: bg-none
:width: 60%
:align: center
```
Figure Q11s: Double U-tube manometer showing equilibrium level choices for solution (not to scale)

In the far right hand arm at level 1

$p_1=p_{\mathrm{atm}}$

and at level 2 

$p_2=p_1+\rho_wgh_1 = p_{\mathrm{atm}}+\rho_wgh_1 \quad\dots(\ast)$. 

In the second limb from right, level 2 also exists within the water layer and $p_2'=p_2$. In the  second and third limbs from the right, level 2 exists in air and in both. Hence $p_2''=p_2'=p_2$.

Level 3 connects the mercury in the third from right limb to the left hand limb with 

$p_3=p_2''+\rho_mgh_2$ and 

$p_3'=p_A+\rho_wgh_3$. With $p_3=p_3'$

$p_A=p_2''+\rho_mgh_2 - \rho_wgh_3 = p_2+\rho_mgh_2 - \rho_wgh_3$.

Eliminating $p_2$ from ($\ast$)

$p_A = p_{\mathrm{atm}}+ g(\rho_wh_1 +\rho_mh_2 - \rho_wh_3)$

With

$ p_{\mathrm{atm}}=1.01\times10^5\,\mathsf{Pa}, \qquad \rho_{\mathrm{Hg}}=13600\,\mathsf{kg}/\mathsf{m}^3 $

Hence,

$ \underline{p_A=124.7\,\mathsf{kPa}} $

::::


:::{exercise} 📝

An inverted U-tube manometer connects pipes A and B as shown in Figure Q12. The manometer fluid is oil of density $900\,\mathsf{kg}/\mathsf{m}^3$; pipes A and B contain water.

```{figure} images/inv_man.png
:alt: Inverted U-tube manometer
:class: bg-none
:width: 40%
:align: center
```
Figure Q12: Inverted U-tube manometer with readings, for use with exercise 12 (not to scale)

Calculate $p_A-p_B$, where $p_A$ and $p_B$ are the pressures at the pipe centres, as shown.

::::{hint} Hint
:class: dropdown
Think about what you are required to calculate. Can you calculate directly?
::::

::::{dropdown} ✓ Check your answer
$-4.218\,\mathsf{kPa}$
::::

::::{admonition} Solution
:class: dropdown
The key to this question is to realise that you are calculating the difference between pressures at A and B.

So we chose points in the system that allow us to understand/relate the pressures at A and B. We know that the pressure within the same fluid is constant at the same horizontal plane. So we analyse first the left hand arm, and then the right hand arm at the level of the top manometer fluid interface. 

We need the difference between the pressures at A and B, so there is no need to calculate either pressure individually.

```{figure} images/inv_man_soln.png
:alt: Inverted U-tube manometer
:class: bg-none
:width: 40%
:align: center
```
Figure Q12s: Inverted U-tube manometer showing equilibrium level choices for solution (not to scale)


In the left hand limb

$p_1 = p_A - \rho_w g h_1$. $\quad\dots(\ast)$

In the right hand limb, the pressure at that level $p_1'=p_1$, by equilibrium. But we can also write this pressure in terms of $p_B$, 

$p_B = p_1 + \rho_ogh_2 + \rho_wgh_3$. 

Eliminating $p_1$ using $(\ast)$

$p_B = p_A - \rho_w g h_1 + \rho_ogh_2 + \rho_wgh_3$ and

$p_A-p_B = \rho_w g h_1 - \rho_ogh_2 - \rho_wgh_3$

Using $\rho_w=1000,\mathsf{kg}/\mathsf{m}^3$ and $\rho_{\mathrm{oil}}=900,\mathsf{kg}/\mathsf{m}^3$, with $h_1=0.45\,\mathsf{m}$, $h_2=0.2\,\mathsf{m}$ and $h_3=0.7\,\mathsf{m}$, 

$ \underline{p_A-p_B=-4.218\,\mathsf{kPa}}.$

::::

:::{exercise} 📝

Figure Q13 shows two enclosed tanks, A and B, which are partially filled with a liquid of relative density $0.82$.  The free surface in tank A is $2\,\mathsf{m}$ below that in tank B. The pressures at the corresponding free surfaces are $p_A$ and $p_B$ with $p_A > p_B$.  The tanks are connected to a u-tube manometer containing mercury of relative density 13.56 (density relative to water).  

```{figure} images/man_tanks.png
:alt: Tanks conected by a manometer
:class: bg-none
:width: 40%
:align: center
```

Determine the pressure difference $p_A - p_B$ when the difference in mercury levels is $225\,\mathsf{mm}$.

::::{dropdown} ✓ Check your answer
$44.209\,\mathsf{kPa}$
::::

::::{admonition} Solution
:class: dropdown
The trick here is to realise that you can analyse the left hand and right hand arms of the system, allowing you to directly relate the difference between the pressures $p_A$ and $p_B$.

```{figure} images/man_tanks_soln.png
:alt: Tanks conected by a manometer
:class: bg-none
:width: 40%
:align: center
```
Figure Q13s: Enclosed tanks connected by a manometer with readings


$p_x = p_A+\rho_1gz_1 = p_B+(z_2+2)\rho_1g+0.225\rho_mg$

Rearranging for the pressure difference

$p_A-p_B = 0.225\rho_mg+(z_2-z_1+2)\rho_1g)$

$\quad = g[0.225\rho_m+(2-0.225)\rho_1]$. 

But we know $\rho_m$ and $\rho_1$ as their relative densities, hence

$p_A-p_B = \rho_wg[0.225\Rho_m + 1.775\Rho_1] = 1000\times 9.81[0.225\times 13.6 + 1.775\times 0.82] $

$\implies \underline{p_A-p_B=44.209\,\mathsf{kPa}}$

::::
